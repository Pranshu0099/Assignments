###**Question 1:** What is hypothesis testing in statistics?
###Answer:

A hypothesis is an assumption about a population parameter, such as the population mean or proportion. Since collecting data from every member of a population is often impractical, hypothesis testing provides a scientific way to draw conclusions based on a representative sample.

Hypothesis testing is a statistical method used to determine whether there is enough evidence from a sample to support or reject a claim about a population. It helps researchers and analysts make decisions using sample data instead of studying the entire population.



### Steps in Hypothesis Testing

1. State the null hypothesis (H₀).
2. State the alternative hypothesis (H₁).
3. Select the significance level (α), usually 0.05.
4. Collect sample data.
5. Calculate the appropriate test statistic (Z-test or T-test).
6. Find the p-value or compare the test statistic with the critical value.
7. Reject or fail to reject the null hypothesis.
8. Draw the final conclusion.

### Example

Suppose a company claims that the average lifetime of its LED bulbs is *10,000 hours*.

* *H₀:* μ = 10,000 hours
* *H₁:* μ ≠ 10,000 hours

A random sample of bulbs is tested. Based on the statistical analysis, the company either rejects or fails to reject the claim.

### Applications

* Medical research
* Business analytics
* Manufacturing quality control
* Finance
* Machine learning
* Market research

### Advantages

* Provides objective decision-making.
* Reduces personal bias.
* Supports evidence-based conclusions.
* Widely applicable across industries.

###**Question 2:** What is the null hypothesis, and how does it differ from the alternative hypothesis?
###Answer:

In hypothesis testing, two competing hypotheses are formulated:

* *Null Hypothesis (H₀)*
* *Alternative Hypothesis (H₁ or Ha)*

## Null Hypothesis (H₀)

The null hypothesis is the default assumption that there is *no significant difference, **no effect, or **no relationship* between variables. It is assumed to be true unless the sample data provides sufficient evidence against it.

*Example:*

A pharmaceutical company claims that a new medicine does not change patients' average blood pressure.

H₀: μ = 120 mmHg


## Alternative Hypothesis (H₁)

The alternative hypothesis states that there *is a significant difference, **effect, or **relationship*.

Example:

H₁: μ ≠ 120 mmHg

If enough statistical evidence exists, the null hypothesis is rejected in favor of the alternative hypothesis.

### Differences Between H₀ and H₁

| Basis              | Null Hypothesis (H₀)    | Alternative Hypothesis (H₁)       |
| ------------------ | ----------------------- | --------------------------------- |
| Meaning            | No difference or effect | Difference or effect exists       |
| Symbol             | H₀                      | H₁ or Ha                          |
| Initial Assumption | Assumed true            | Considered only if H₀ is rejected |
| Goal               | Tested directly         | Supported indirectly              |

The null hypothesis represents the existing assumption of no effect, while the alternative hypothesis represents the research claim that an effect or difference exists.

### Real-Life Example

A retail company introduces a new advertising campaign.

* H₀: The campaign does not increase average sales.
* H₁: The campaign increases average sales.

After analyzing sales data, management decides whether there is sufficient evidence to conclude that the campaign improved sales.

##**Question 3:** Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.
##Answer:

The *significance level, represented by the symbol **α (alpha), is the maximum probability of making a **Type I error*, which occurs when a true null hypothesis is incorrectly rejected.

The significance level acts as a threshold for deciding whether the evidence obtained from a sample is strong enough to reject the null hypothesis.

The most commonly used significance levels are:

* α = 0.05 (5%)
* α = 0.01 (1%)
* α = 0.10 (10%)

Among these, *0.05* is the most widely used in research and business analytics.

## Role of Significance Level in deciding outcome of a test

1. Determines the rejection region.
2. Helps compare the p-value with α.
3. Controls the probability of Type I error.
4. Ensures consistency in statistical decision-making.

### Decision Rule

* If *p-value ≤ α*, reject the null hypothesis.
* If *p-value > α*, fail to reject the null hypothesis.

### Example

Suppose a researcher tests whether a new teaching method improves student performance.

* Significance level = 0.05
* Calculated p-value = 0.03

Since *0.03 < 0.05*, the null hypothesis is rejected, indicating sufficient evidence that the teaching method improves performance.

### Importance

* Controls false positive results.
* Improves reliability of research.
* Provides an objective basis for decision-making.
* Widely used in medicine, engineering, economics, and data science.


The significance level is a critical parameter in hypothesis testing because it determines the threshold for rejecting the null hypothesis. It balances the risk of making incorrect conclusions while ensuring statistically sound decisions



###**Question 4:** What are Type I and Type II errors? Give examples of each.
##Answer:

In hypothesis testing, decisions are made based on sample data. Since these decisions are based on samples rather than the entire population, errors may occur. The two main types of errors are *Type I Error* and *Type II Error*.

### 1. Type I Error (False Positive)

A *Type I Error* occurs when the *null hypothesis (H₀) is true, but it is incorrectly rejected*.

In simple words, we conclude that a significant effect exists when, in reality, it does not.

The probability of committing a Type I Error is denoted by *α (alpha)*, also known as the significance level.

#### **Example**

Suppose a pharmaceutical company tests a new medicine.

* *H₀:* The medicine has no effect.
* *H₁:* The medicine is effective.

If the medicine actually has no effect but the researcher concludes that it is effective, a *Type I Error* has occurred.

### 2. Type II Error (False Negative)

A *Type II Error* occurs when the *null hypothesis is false, but it is not rejected*.

In other words, we conclude that there is no significant effect even though an effect actually exists.

The probability of a Type II Error is represented by *β (beta)*.

####** Example**


Using the same medicine example:

If the medicine actually works but the researcher concludes that it does not work, a *Type II Error* has occurred.


## Differences Between Type I and Type II Errors

| Basis       | Type I Error                  | Type II Error             |
| ----------- | ----------------------------- | ------------------------- |
| Decision    | Reject a true H₀              | Fail to reject a false H₀ |
| Also Called | False Positive                | False Negative            |
| Probability | α (Significance Level)        | β                         |
| Result      | Incorrectly detects an effect | Misses a real effect      |


## Real-Life Examples

### Medical Diagnosis

* *Type I Error:* A healthy person is diagnosed as having a disease.
* *Type II Error:* A person with the disease is diagnosed as healthy.

### Email Spam Filter

* *Type I Error:* A genuine email is marked as spam.
* *Type II Error:* A spam email reaches the inbox.

### Manufacturing

* *Type I Error:* A good product is rejected during inspection.
* *Type II Error:* A defective product is accepted and sent to customers.


##**Question 5:** What is the difference between a Z-test and a T-test? Explain when to use each
##Answer:

Both the *Z-test* and *T-test* are statistical methods used to test hypotheses about population means. The choice between them depends mainly on the sample size and whether the population standard deviation is known.

### Z-Test

A *Z-test* is used when:

* The sample size is *30 or more*.
* The population standard deviation (*σ*) is known.
* The data follows a normal distribution or the sample is sufficiently large.

#### Applications

* Comparing sample means with population means.
* Comparing the means of two large independent samples.
* Testing population proportions.

### T-Test

A *T-test* is used when:

* The sample size is *less than 30*.
* The population standard deviation is *unknown*.
* The sample is drawn from a normally distributed population.

The T-test uses the *Student's t-distribution*, which accounts for additional uncertainty in small samples.

### Difference Between Z-Test and T-Test

| Basis                         | Z-Test                       | T-Test                     |
| ----------------------------- | ---------------------------- | -------------------------- |
| Sample Size                   | Large (n ≥ 30)               | Small (n < 30)             |
| Population Standard Deviation | Known                        | Unknown                    |
| Distribution Used             | Standard Normal Distribution | Student's t-Distribution   |
| Accuracy                      | Suitable for large samples   | Suitable for small samples |

---

## Example

Suppose a company wants to determine whether the average weight of packaged products differs from the standard weight.

* If *100 products* are sampled and the population standard deviation is known, a *Z-test* is appropriate.
* If *20 products* are sampled and the population standard deviation is unknown, a *T-test* should be used.

```python
'''
Question 7: Implement hypothesis testing using Z-statistics for a sample
dataset in Python. Show the Python code and interpret the results.
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2,
49.6, 50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
'''
#Answer Code


import numpy as np
import scipy.stats as stats

# Data
sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
    50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
    50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
    50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

# Parameters
mu_0 = 50.0  # Hypothesized population mean
alpha = 0.05  # Significance level

# Descriptive Statistics
n = len(sample_data)
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)  # Sample standard deviation (s)
standard_error = sample_std / np.sqrt(n)

# Z-Test Calculation
z_score = (sample_mean - mu_0) / standard_error
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))  # Two-tailed p-value

# Output
print(f"Sample Size (n): {n}")
print(f"Sample Mean (x̄): {sample_mean:.4f}")
print(f"Sample Std Dev (s): {sample_std:.4f}")
print(f"Standard Error (SE): {standard_error:.4f}")
print(f"Calculated Z-Score: {z_score:.4f}")
print(f"P-Value: {p_value:.4f}")

# Decision Logic
z_critical = stats.norm.ppf(1 - alpha / 2)
if abs(z_score) > z_critical:
    print(f"\nDecision: Reject H0 (Z = {z_score:.2f} lies in the rejection region)")
else:
    print(f"\nDecision: Fail to reject H0 (Z = {z_score:.2f} is within the acceptance region)")

    x = np.linspace(-3, 3, 100)
y = stats.norm.pdf(x, 0, 1)

plt.plot(x, y, label="Normal Distribution")
plt.axvline(z_score, color="green", label=f"Our Z-Score ({z_score:.2f})")
plt.axvline(1.96, color="red", linestyle="--", label="Critical Bound (+1.96)")
plt.axvline(-1.96, color="red", linestyle="--", label="Critical Bound (-1.96)")

plt.title("Z-Test Result")
plt.legend()
plt.show()

```

##Interpretation of Results:

1. Since caliculated z-scores(0.9940) is less than the critical bounds(1.96) value, the Z-statistic falls well inside the non-rejection (acceptance) region.
2. Since $p = 0.3202 > \alpha = 0.05$, there is a $32.02\%$ probability of observing a sample mean this far from $50.0$ purely due to random sampling variation.

Hence, we fail to reject the null hypothesis ($H_0$). There is no statistically significant evidence at the $5\%$ level to suggest that the true population mean differs from $50.0$


```python
'''
Question 8: Write a Python script to simulate data from a normal distribution
and calculate the 95% confidence interval for its mean. Plot the data using
Matplotlib
'''
#Anwer code

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Generate simulated dataset
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)

# Calculate statistics
mean = np.mean(data)
se = stats.sem(data)

# 95% Confidence interval
ci_lower, ci_upper = stats.t.interval(0.95, df=len(data)-1, loc=mean, scale=se)
ci_clean = (round(float(ci_lower), 2), round(float(ci_upper), 2))

print("Mean:", round(mean, 2))
print("95% CI:", ci_clean)

# Simple plot
plt.hist(data, bins=15, alpha=0.7)
plt.axvline(mean, color='black', label='Mean')
plt.axvline(ci_lower, color='red', linestyle='--', label='95% CI')
plt.axvline(ci_upper, color='red', linestyle='--')

plt.xlabel("Values")
plt.ylabel("Count")
plt.legend()
plt.show()
```

```python
'''
Question 9: Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores
represent in terms of standard deviations from the mean.
'''
#Answer Code

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Sample dataset
data = [45, 50, 52, 47, 49, 53, 51, 48, 50, 46]

# Calculate Z-scores
z_scores = zscore(data)

print("Original Data:")
print(data)

print("\nZ-Scores:")
print(np.round(z_scores,2))

# Plot Histogram
plt.figure(figsize=(8,5))
plt.hist(z_scores,
         bins=6,
         edgecolor='black')

plt.title("Histogram of Standardized Data (Z-Scores)")
plt.xlabel("Z-Score")
plt.ylabel("Frequency")
plt.grid(alpha=0.3)

plt.show()
```
