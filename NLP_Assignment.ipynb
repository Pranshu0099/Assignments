**Question 1:What is Natural Language Processing? Explain any four real-world applications of NLP. Why can a machine not directly understand raw text? Describe the main stages of a basic NLP pipeline.**

**Answer:** Natural Language Processing (NLP) is a subfield of artificial intelligence, computer science, and linguistics focused on enabling computers to understand, interpret, process, and generate human languages. The ultimate goal is to bridge the communication gap between human language (unstructured, nuanced, and ambiguous) and computer code (structured, deterministic, and mathematical).

**Real-World Applications of NLP**

1. Machine Translation: Automatically translating written or spoken content from one language to another (e.g., Google Translate, DeepL) while preserving context, grammatical structure, and idiomatic meaning.

2. Sentiment Analysis: Analyzing user-generated text—such as product reviews, social media posts, or customer feedback—to classify the underlying emotional tone as positive, negative, or neutral.

3. Conversational AI and Virtual Assistants: Powering chatbots, customer service automation agents, and voice assistants (e.g., Siri, Alexa) to parse user intents, extract relevant entities, and generate context-aware responses.

4. Information Extraction and Named Entity Recognition (NER): Scanning unstructured documents (contracts, medical records, financial reports) to locate and classify key elements into predefined categories like person names, organizations, dates, monetary figures, and locations.

**Why Machines Cannot Directly Understand Raw Text**

Computational Nature: Computers are binary calculating engines designed to perform mathematical operations on numerical arrays and matrices, not discrete strings or characters.

Semantic Ambiguity and Polysemy: Words have different meanings depending on context (e.g., "bank" can refer to a financial institution or the side of a river; "apple" can mean a fruit or a corporation).

Syntax and Grammar Variability: Human languages have irregular rules, complex sentence structures, idioms, sarcasm, slang, typos, and cultural context that cannot be parsed by hardcoded programmatic rules.

Lack of Spatial/Relational Metric: Characters and strings have no intrinsic numerical relationship—a machine has no native way of knowing that the string "king" is semantically closer to "queen" than to "refrigerator" without vector representations.

**Main Stages of a Basic NLP Pipeline**

1. Data Ingestion & Preprocessing (Text Normalization):

Sentence Segmentation, Tokenization, Lowercasing, Noise Removal, Stop-Word Removal.

2. Morphological Analysis & Normalization

Stemming, Lemmatization

3. Syntactic & Semantic Analysis (Feature Extraction)

Part-of-Speech (POS) Tagging, Named Entity Recognition (NER), Dependency Parsing

4. Vectorization (Word & Document Representation)

Converting clean tokens into numerical vectors that algorithms can compute Classical approaches, Dense semantic embeddings

5. Modeling & Downstream Tasks

Feeding the numeric vectors into a machine learning or deep learning model

Describe the following operations with suitable examples:

**Answer:** **Purpose of Text Preprocessing in NLP**

Text preprocessing is the foundational phase in Natural Language Processing where raw, noisy, and unstructured textual data is cleaned, standardized, and transformed into a structured format suitable for machine learning models.

Noise Reduction: Removing non-informative elements (like random punctuation, formatting artifacts, or HTML tags) that do not carry semantic meaning.

Vocabulary Reduction: Consolidating different surface forms of the same word (e.g., "Run", "running", and "runs") into a single base representation to shrink the feature space.

Sparsity Mitigation: Preventing high-dimensional, sparse matrices in vectorization steps (such as Bag-of-Words and TF-IDF).

Improving Model Accuracy & Efficiency: Ensuring the learning algorithm focuses on meaningful linguistic patterns rather than surface-level variations or formatting quirks.

1. Lowercasing

    Description: Converting all alphabetic characters in the text to lowercase. This ensures that the model treats words identically regardless of their capitalization or position in a sentence.

    Example:

        Input: "The quick Brown FOX jumps."

        Output: "the quick brown fox jumps."

2. Removing Punctuation

    Description: Stripping out symbols such as periods, commas, exclamation marks, quotation marks, and brackets. In most standard classification and retrieval tasks, punctuation marks add noise without contributing significant semantic value.

    Example:

        Input: "Hello, world! How's everything going?"

        Output: "Hello world Hows everything going"


3. Removing Numbers

    Description: Eliminating numeric digits from the text when numerical quantities do not provide relevant predictive features for the target downstream task (e.g., generic sentiment analysis).

    Example:

        Input: "I bought 3 laptops and 2 phones in 2024."

        Output: "I bought  laptops and  phones in ."


4. Removing Stopwords

    Description: Filtering out high-frequency structural words (such as "the", "is", "at", "which", "on", "and") that appear abundantly across documents but convey minimal distinctive semantic content.

    Example:

        Input: "This is an example showing the removal of stopwords."

        Output: "example showing removal stopwords"

5. Handling Extra Whitespaces

    Description: Collapsing multiple consecutive spaces, leading/trailing spaces, newlines (\n), and tabs (\t) into a single uniform space delimiter.

    Example:

        Input: "  Natural   Language\t\tProcessing   \n  "

        Output: "Natural Language Processing"

6. Tokenization

    Description: The process of segmenting a continuous string of text into discrete linguistic units called tokens (such as words or sentences).

    Example (Word Tokenization):

        Input: "Deep learning is powerful."

        Output: ["Deep", "learning", "is", "powerful", "."]

7. Stemming

    Description: A fast, rule-based heuristic process that chops off suffixes and prefixes (affixes) from words to reduce them to a common base or root form. Because it does not use a dictionary, the resulting root ("stem") may not be a valid dictionary word.

    Common Algorithm: Porter Stemmer, Snowball Stemmer.

    Example:

        Input Words: ["studies", "studying", "studied", "generously"]

        Output Stems: ["studi", "studi", "studi", "gener"]

8. Lemmatization

    Description: An advanced morphological analysis technique that utilizes a vocabulary dictionary and Part-of-Speech (POS) tags to reduce a word to its valid canonical base form (known as a lemma). Unlike stemming, lemmatization always produces a valid dictionary word.

    Example:

        Input Words: ["better", "running", "was", "mice"]

        Output Lemmas: ["good", "run", "be", "mouse"]


9. Unigrams, Bigrams, and Trigrams (N-grams)Description: Contiguous sequences of $n$ items (typically words) extracted from a given sample of text. They help capture local word order and multi-word contextual phrases that individual words miss.

**Given Sentence: "The cat sat on the mat"

Unigrams ($n=1$): Sequences of single words. Output: ["The", "cat", "sat", "on", "the", "mat"]

Bigrams ($n=2$): Sequences of two adjacent words. Output: ["The cat", "cat sat", "sat on", "on the", "the mat"]

Trigrams ($n=3$): Sequences of three adjacent words. Output: ["The cat sat", "cat sat on", "sat on the", "on the mat"]

**Answer:** NLTK (Natural Language Toolkit)

    Core Characteristics: Provides a massive collection of corpora, lexical resources (like WordNet), and numerous implementations of classic NLP algorithms (multiple stemmers, parsers, and chunkers). It prioritizes modularity and flexibility over raw execution speed.

spaCy

    Main Purpose: An industrial-strength, production-grade NLP library engineered for high performance, efficiency, and real-world software applications.

    Core Characteristics: Implemented in optimized C/Cython. It offers batteries-included, pre-trained neural pipeline components (deep learning-based POS taggers, dependency parsers, and Named Entity Recognizers) that execute end-to-end processing rapidly.

    Common Use Cases: Large-scale data extraction pipelines, fast production Named Entity Recognition (NER), dependency parsing for relationship extraction, and integrated preprocessing for deep learning frameworks.

Gensim

    Main Purpose: A specialized framework dedicated specifically to unsupervised topic modeling, document indexing, and similarity retrieval via vector space algorithms.

    Core Characteristics: Highly optimized for handling large text corpora without loading everything into memory (stream processing). It contains efficient C-accelerated implementations of Word2Vec, FastText, Doc2Vec, Latent Dirichlet Allocation (LDA), and Latent Semantic Analysis (LSA).

    Common Use Cases: Discovering latent topics across document collections (Topic Modeling), training custom word/document embeddings, and computing semantic similarity between large text documents.

**What Are Word Embeddings?**

Word embeddings are real-valued, continuous vector representations of words in a predefined multi-dimensional space (typically 50 to 300 dimensions).Instead of treating words as isolated discrete symbols, embeddings place semantically similar words close to one another in vector space based on the distributional hypothesis ("words that occur in similar contexts tend to have similar meanings"). This geometric representation allows mathematical relationships to encode semantic relationships (e.g., $\vec{v}_{\text{King}} - \vec{v}_{\text{Man}} + \vec{v}_{\text{Woman}} \approx \vec{v}_{\text{Queen}}$).

**How Dense Word Embeddings Differ from Bag-of-Words (BoW) and TF-IDF**

1. Dimensionality and Sparsity

BoW & TF-IDF (Sparse): The vector length equals the entire size of the corpus vocabulary ($\vert{}V\vert{}$), which often spans $20{,}000$ to $100{,}000+$ dimensions. Each document or word vector is extremely sparse, containing mostly zeros and a few non-zero counts/weights.

Dense Embeddings: Vectors have a fixed, compact dimension (e.g., 100, 300, or 768 dimensions) regardless of vocabulary size. Every entry in the vector is a continuous floating-point number, packing rich information into a dense array.

2. Semantic and Contextual Understanding

BoW & TF-IDF (No Semantics): Treat every word as completely orthogonal and independent. The words "automobile" and "car" receive completely different indices and have a dot product of zero, meaning the model sees no intrinsic semantic similarity between them.

Dense Embeddings: Learn semantic proximity through surrounding context. The vectors for "automobile" and "car" will have a high cosine similarity score (close to $1.0$), allowing downstream models to generalize better across synonyms.

3. Word Order and Positional Context

BoW & TF-IDF: Discard grammatical structure and word order entirely (treating text as an unordered bag of frequency counts).

Dense Embeddings (e.g., Word2Vec Skip-gram/CBOW, GloVe): Are trained explicitly by learning local sliding-window contextual co-occurrence patterns, capturing syntactic and semantic dependencies.

4. Out-of-Vocabulary (OOV) and Scalability

BoW & TF-IDF: As new documents and words arrive, the vocabulary dimension grows linearly, leading to the curse of dimensionality and increased memory usage.

Dense Embeddings: The vector dimension remains constant. Modern subword embeddings (like FastText) can even infer vectors for unseen words by decomposing them into character n-grams.

**Question 4:What is sequential data? Explain why an ordinary feed-forward neural network is not
sufficient for many NLP tasks.
Compare RNN, LSTM, and GRU. Also explain accuracy, precision, recall, F1-score, and
perplexity**

**Answer:** Sequential data is any data where the order, timing, or relative position of individual elements matters as much as the elements themselves. In a sequence, each data point cannot be evaluated in total isolation because its meaning and function depend on preceding and succeeding elements.

***Why Ordinary Feed-Forward Neural Networks (FFNNs) Fail for Many NLP Tasks***

1. Inability to Handle Variable-Length Inputs: Standard FFNNs (e.g., Dense/MLP layers) require fixed-size input vectors. Sentences and paragraphs naturally vary in length from a few words to hundreds of words.

2. Destruction of Sequential and Positional Context: When an FFNN takes an input (like a flattened Bag-of-Words or concatenated vector), it treats features as static spatial attributes. It has no intrinsic concept of sequential progression or relative word order.

3. Absence of Internal Memory / State: FFNNs process each input independently with zero temporal persistence. They cannot carry context learned at step $t=1$ forward to resolve dependencies at step $t=20$ (e.g., matching a pronoun to its antecedent across a long sentence).

4. Parameter Inefficiency: If you force a fixed maximum window size on an FFNN to handle sequences, every word position needs separate weight connections, causing explosive parameter growth without sharing learned features across different positions.

**Comparison of RNN, LSTM, and GRU**

1. Standard RNN (Recurrent Neural Network)

Core Idea: Processes inputs step-by-step ($x_t$), maintaining a single hidden state ($h_t$) that passes information from the previous time step to the current one via a cyclic loop:$$h_t = \tanh(W_{hh} h_{t-1} + W_{xh} x_t + b)$$

Advantages: Simple architecture, few parameters, computationally lightweight.

Limitations: Suffers heavily from vanishing and exploding gradients during Backpropagation Through Time (BPTT). It cannot maintain dependencies across long sequences ($>10$ steps).

2. LSTM (Long Short-Term Memory)

Core Idea: Explicitly controls memory through a dedicated Cell State ($C_t$) (the long-term memory highway) regulated by three multiplicative gating mechanisms:

> Forget Gate ($f_t$): Decides what information to discard from the previous cell state.

> Input Gate ($i_t$): Decides what new candidate information ($\tilde{C}_t$) to store in the cell state.

> Output Gate ($o_t$): Decides what parts of the updated cell state to emit as the hidden state ($h_t$).

Advantages: Effectively solves the vanishing gradient problem; excels at capturing long-range dependencies across hundreds of time steps.

Limitations: High computational complexity with 4 separate linear transformations per step, requiring more memory and longer training times.

3. GRU (Gated Recurrent Unit)

Core Idea: A streamlined, simplified variant of the LSTM that merges the cell state and hidden state into a single Hidden State ($h_t$) using only two gates:

> Reset Gate ($r_t$): Decides how much of past memory to forget when computing new candidate state.

> Update Gate ($z_t$): Simultaneously balances how much past information to keep versus how much new candidate state to incorporate (acting as both forget and input gates combined).

Advantages: Faster training and lower memory footprint than LSTM (fewer parameters) while achieving comparable performance on most sequence modeling tasks.

Limitations: Lacks a strictly separate long-term cell state channel, which can slightly reduce performance on extremely long, complex sequences compared to carefully tuned LSTMs.

**Evaluation Metrics**

1. Accuracy: The ratio of correct predictions (both positive and negative) to total predictions.$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$When to Use: Balanced datasets. It is misleading for imbalanced tasks (e.g., if 99% of spam filters see ham, a dummy classifier guessing "ham" gets 99% accuracy).

2. Precision: Out of all instances predicted as positive, how many were actually positive?$$\text{Precision} = \frac{TP}{TP + FP}$$Focus: Minimizing False Positives (FP). Crucial in scenarios like spam filtering or content moderation (where falsely flagging legitimate content is costly).

3. Recall (Sensitivity): Out of all actual positive instances, how many did the model correctly identify?$$\text{Recall} = \frac{TP}{TP + FN}$$Focus: Minimizing False Negatives (FN). Crucial in medical diagnosis or fraud detection (where missing an actual positive case has severe consequences).

4. F1-Score: The harmonic mean of Precision and Recall, providing a balanced single metric when there is an uneven class distribution:$$\text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} = \frac{2TP}{2TP + FP + FN}$$Focus: Evaluates models effectively on imbalanced datasets where both false positives and false negatives carry significant weight.

5. Perplexity (PPL): The standard intrinsic evaluation metric for Language Models (LMs). It measures how well a probability distribution or language model predicts a sample of text. Mathematically, it is the exponentiated average negative log-likelihood (cross-entropy loss $\mathcal{L}$) per token:$$\text{Perplexity} = \exp\left(-\frac{1}{N}\sum_{i=1}^{N} \log P(w_i \mid w_1, \dots, w_{i-1})\right) = e^{\mathcal{L}}$$Interpretation: Perplexity represents the effective "branching factor" of the model—i.e., the number of equally likely words the model is choosing between at each step. A lower perplexity indicates a better, more confident language model.

**Question 5: What is a Sequence-to-Sequence model? Explain the roles of the encoder,
decoder, hidden states, and context vector.
Why does a basic Seq2Seq model struggle with long sentences? Explain how the
attention mechanism solves this problem.**

**Answer:** A Sequence-to-Sequence (Seq2Seq) model is a deep learning architecture designed to transform an input sequence from one domain into an output sequence in another domain, where the input and output lengths can differ dynamically.

A basic Seq2Seq model follows an Encoder-Decoder framework, commonly built using recurrent units (RNN, LSTM, or GRU).

1. Encoder:Reads the input sequence $(x_1, x_2, \dots, x_T)$ step-by-step.At each time step $t$, it consumes the current token embedding and the previous hidden state to compute a new hidden state.Its primary objective is to encode the full semantic meaning of the source sequence into an internal representation.

2. Hidden States ($h_t$ and $s_t$):Encoder Hidden States ($h_t$): The memory vectors emitted at every time step $t$ representing local contextual information up to that token.Decoder Hidden States ($s_t$): The working memory of the decoder at step $t$, tracking both what has already been generated and what remains to be emitted.

3. Context Vector ($c$):In the standard Seq2Seq model, the final hidden state of the encoder ($h_T$) serves as the single Context Vector ($c = h_T$).It acts as the informational bridge passed from the encoder to the decoder, supposedly encapsulating the meaning of the entire source sentence.

4. Decoder:Takes the context vector $c$ (often as its initial hidden state $s_0$) and generates the target sequence token-by-token.At each step, it takes the previously generated token $y_{t-1}$ and its current hidden state $s_t$ to produce a probability distribution over the target vocabulary via a Softmax layer, continuing until it emits an end-of-sequence token (<EOS>).

**Why Basic Seq2Seq Models Struggle with Long Sentences**

1. Fixed-Length Vector Compression: The encoder is forced to compress an entire sentence—regardless of whether it contains 5 words or 60 words—into a single, static, fixed-dimension context vector $c$.

2. Information Loss & Recency Bias: As the recurrent encoder processes very long sentences, earlier words are overwritten and diluted by later words due to the sequential nature of RNN/LSTM updates.

3. Degraded Decoding: The decoder receives only one global summary vector at step 0. By step 15 or 20 of decoding, the initial context signal fades, causing grammatical degradation, hallucinations, or truncated translations.

**How the Attention Mechanism Solves the Bottleneck**

The Attention Mechanism eliminates the static single-vector bottleneck by creating a dynamic direct connection between the decoder and all intermediate encoder hidden states.

1. Retaining All Encoder Hidden States:Instead of discarding intermediate states $(h_1, h_2, \dots, h_T)$, the model stores every encoder hidden state in memory.

2. Dynamic Alignment Scores (Attention Scores):At each decoder step $t$, the model calculates a compatibility score $e_{t, i}$ between the current decoder hidden state $s_{t-1}$ and each source encoder state $h_i$:$$e_{t, i} = \text{score}(s_{t-1}, h_i)$$

3. Soft Attention Weights ($\alpha_{t, i}$):The scores are normalized across all input positions using a Softmax function, producing a probability distribution that sums to $1.0$:$$\alpha_{t, i} = \frac{\exp(e_{t, i})}{\sum_{k=1}^{T} \exp(e_{t, k})}$$These attention weights quantify exactly how much "focus" the decoder should place on input word $i$ when predicting output word $t$.


4. Dynamic, Time-Varying Context Vector ($c_t$):The decoder computes a distinct, customized context vector $c_t$ at each decoding step as a weighted sum of the encoder hidden states:$$c_t = \sum_{i=1}^{T} \alpha_{t, i} h_i$$

5. Contextual Generation:The decoder combines $s_t$ and the specific $c_t$ to predict the next word. When translating a specific word (e.g., translating the 4th word of a 30-word sentence), the network shifts high attention weight directly to the corresponding source region, eliminating memory decay and dramatically boosting performance on long sequences.

```python
'''
Question 6: Building a Reusable Text-Preprocessing Function
Write a Python program to clean the following text:
" This PRODUCT is Amazing!!! I bought 2 items, and I really loved them. "
Your program should:
● Convert the text to lowercase.
● Remove punctuation.
● Remove numbers.
● Remove extra whitespaces.
● Remove English stopwords.
● Tokenize the cleaned text.
● Create a reusable preprocessing function
'''

import re
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download necessary NLTK datasets
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)


def preprocess_text(text: str) -> list[str]:
    """Cleans and tokenizes raw input text by performing:

    1. Lowercasing
    2. Punctuation removal
    3. Number removal
    4. Extra whitespace normalization
    5. Tokenization
    6. English stopword filtering
    """
    # 1. Convert text to lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # 3. Remove numbers/digits
    text = re.sub(r"\d+", "", text)

    # 4. Remove extra whitespaces (handles multiple spaces, tabs, newlines)
    text = " ".join(text.split())

    # 5. Tokenize the cleaned text
    tokens = word_tokenize(text)

    # 6. Remove English stopwords
    stop_words = set(stopwords.words("english"))
    cleaned_tokens = [word for word in tokens if word not in stop_words]

    return cleaned_tokens


# Sample input string
sample_text = (
    "  This PRODUCT is Amazing!!! I bought 2 items, and I really loved them.  "
)

# Execute the preprocessing pipeline
processed_tokens = preprocess_text(sample_text)

print("Original Text:")
print(f'"{sample_text}"\n')

print("Cleaned Tokens:")
print(processed_tokens)
```

```python
'''
Question 7: Write a Python program using NLTK to:
● Tokenize the sentence below.
● Apply Porter stemming.
● Apply WordNet lemmatization.
● Generate unigrams, bigrams, and trigrams.
Use the sentence:
"The students were studying and running towards their classes."
Compare the stemming and lemmatization outputs.
'''

import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

# Download required NLTK resources
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

# Input sentence
sentence = "The students were studying and running towards their classes."

# 1. Tokenization
tokens = word_tokenize(sentence)

# 2. Porter Stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(token) for token in tokens]

# 3. WordNet Lemmatization (using verb POS tag for action words, noun for others)
lemmatizer = WordNetLemmatizer()
# Lemmatizing with explicit POS tags to show accurate base forms
lemmatized_tokens = [
    lemmatizer.lemmatize(
        token.lower(),
        pos="v"
        if token.lower() in ["were", "studying", "running"]
        else "n"
        if token.lower() in ["students", "classes"]
        else "n",
    )
    for token in tokens
]

# 4. Generate Unigrams, Bigrams, and Trigrams
# Filtering punctuation tokens for clean n-grams
clean_tokens = [t for t in tokens if t.isalnum()]

unigrams = list(ngrams(clean_tokens, 1))
bigrams = list(ngrams(clean_tokens, 2))
trigrams = list(ngrams(clean_tokens, 3))

# Display Outputs
print("Original Sentence:")
print(f'"{sentence}"\n')

print("1. Tokenized Words:")
print(tokens)
print()

print("2. Porter Stemmed Tokens:")
print(stemmed_tokens)
print()

print("3. WordNet Lemmatized Tokens:")
print(lemmatized_tokens)
print()

print("4. N-Grams:")
print(f"Unigrams: {unigrams}\n")
print(f"Bigrams:  {bigrams}\n")
print(f"Trigrams: {trigrams}\n")
```

**Comparison of Stemming vs. Lemmatization Outputs**

"studying"

    Stemmer: Produces "studi" — chops off the -ying suffix using fixed heuristic rules, resulting in a non-dictionary fragment.

    Lemmatizer: Produces "study" — uses dictionary lookup and morphological analysis to return the valid root base verb.

"were"

    Stemmer: Produces "were" — fails to identify irregular verb conjugations because it only looks for common affixes.

    Lemmatizer: Produces "be" — correctly resolves the irregular inflection of the auxiliary verb to its canonical lemma.

"classes" and "students"

    Stemmer: Produces "class" and "student" by stripping the trailing -es and -s.

    Lemmatizer: Produces "class" and "student" by identifying the singular noun forms.

Core Difference:

    Stemming is a faster, rule-based clipping mechanism that often yields non-words (e.g., studi).

    Lemmatization uses linguistic vocabulary rules and Part-of-Speech context to ensure every output is a grammatically valid base word (e.g., study, be).

```python
'''
Question 8: Use the following documents:
Document 1: "NLP is useful for text analysis"
Document 2: "Text analysis is useful for business"
Document 3: "NLP converts text into numerical features"
Write a Python program to:
● Create a Bag of Words representation.
● Display the vocabulary.
● Display the feature matrix.
● Create a TF-IDF representation.
● Display the TF-IDF matrix.
● Identify the word with the highest TF-IDF score in each document
'''

import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Define the corpus of documents
documents = [
    "NLP is useful for text analysis",  # Document 1
    "Text analysis is useful for business",  # Document 2
    "NLP converts text into numerical features",  # Document 3
]

# -------------------------------------------------------------
# 1. Bag of Words (BoW) Representation
# -------------------------------------------------------------
count_vec = CountVectorizer()
bow_matrix = count_vec.fit_transform(documents)
bow_vocab = count_vec.get_feature_names_out()

print("--- 1. Bag of Words (BoW) ---")
print(f"Vocabulary ({len(bow_vocab)} terms):")
print(list(bow_vocab))
print("\nBoW Feature Matrix (Array):")
print(bow_matrix.toarray())

# -------------------------------------------------------------
# 2. TF-IDF Representation
# -------------------------------------------------------------
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(documents)
tfidf_vocab = tfidf_vec.get_feature_names_out()
tfidf_array = tfidf_matrix.toarray()

print("\n--- 2. TF-IDF Representation ---")
print("TF-IDF Matrix (Array, Rounded to 3 decimals):")
print(np.round(tfidf_array, 3))

# -------------------------------------------------------------
# 3. Word with Highest TF-IDF Score in Each Document
# -------------------------------------------------------------
print("\n--- 3. Highest TF-IDF Word per Document ---")
for idx, doc_vector in enumerate(tfidf_array):
    max_idx = np.argmax(doc_vector)
    top_word = tfidf_vocab[max_idx]
    top_score = doc_vector[max_idx]
    print(
        f"Document {idx + 1}: Word = '{top_word}' (Score = {top_score:.4f})"
    )
```

```python
'''
Question 10: Part A: LSTM Model
Build a simple LSTM model in TensorFlow/Keras for binary sentiment classification.
The model should contain:
● A text vectorization or tokenization step
● An Embedding layer
● An LSTM layer
● A Dense output layer
Part B: Attention Calculation
The attention scores for three input words are:
[1.2, 0.5, 2.0]
Use Softmax to calculate the attention weights and identify which input position receives the
highest attention
'''

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# -------------------------------------------------------------
# 1. Sample Dataset (Binary Sentiment: 1 = Positive, 0 = Negative)
# -------------------------------------------------------------
train_sentences = [
    "I loved this movie it was fantastic and brilliant",
    "Great acting and wonderful story truly amazing",
    "Terrible film completely boring and a waste of time",
    "Worst experience ever horrible acting and poor plot",
    "An absolute masterpiece highly recommended",
    "Disappointing and dull I hated every minute",
]
train_labels = np.array([1, 1, 0, 0, 1, 0], dtype=np.int32)

# -------------------------------------------------------------
# 2. Text Vectorization (Preprocessing & Tokenization)
# -------------------------------------------------------------
VOCAB_SIZE = 1000
MAX_LEN = 10

# Create and configure the vectorization layer
vectorize_layer = layers.TextVectorization(
    max_tokens=VOCAB_SIZE,
    output_mode="int",
    output_sequence_length=MAX_LEN,
    standardize="lower_and_strip_punctuation",
)

# Fit vocabulary on training texts
vectorize_layer.adapt(train_sentences)

# Transform raw text into 2D integer tensor (Batch_Size, MAX_LEN)
X_train_indices = vectorize_layer(train_sentences).numpy()

# -------------------------------------------------------------
# 3. Build the LSTM Model (Integer Inputs)
# -------------------------------------------------------------
model = models.Sequential(
    [
        # Accepts integer sequences of fixed length MAX_LEN
        layers.Input(shape=(MAX_LEN,), dtype="int32"),
        # Embedding layer: maps token indices to 32-dim dense vectors
        layers.Embedding(
            input_dim=VOCAB_SIZE, output_dim=32, mask_zero=True
        ),
        # LSTM layer: processes the sequence
        layers.LSTM(units=16, return_sequences=False),
        # Dropout for regularization
        layers.Dropout(0.2),
        # Output layer for binary classification (0 to 1 probability)
        layers.Dense(units=1, activation="sigmoid"),
    ]
)

# -------------------------------------------------------------
# 4. Compile & Train the Model
# -------------------------------------------------------------
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.005),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)

model.summary()

# Train with numerical integer arrays
model.fit(
    X_train_indices,
    train_labels,
    epochs=10,
    batch_size=2,
    verbose=1,
)

# -------------------------------------------------------------
# 5. Predict on New Raw Text
# -------------------------------------------------------------
test_texts = [
    "The movie was brilliant and amazing",
    "Horrible acting and a waste of time",
]

# Vectorize test texts before prediction
test_indices = vectorize_layer(test_texts).numpy()
predictions = model.predict(test_indices)

print("\n--- Predictions ---")
for text, prob in zip(test_texts, predictions.flatten()):
    sentiment = "Positive" if prob >= 0.5 else "Negative"
    print(
        f"Text: '{text}' -> Score: {prob:.4f} ({sentiment})"
    )
```

**Part B: Attention Calculation**

The Softmax function normalizes an array of raw attention scores $z = [z_1, z_2, z_3]$ into attention weights $\alpha = [\alpha_1, \alpha_2, \alpha_3]$:$$\alpha_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$

raw scores: $z_1 = 1.2$, $z_2 = 0.5$, $z_3 = 2.0$

$e^{z_1} = e^{1.2} \approx 3.3201$

$e^{z_2} = e^{0.5} \approx 1.6487$

$e^{z_3} = e^{2.0} \approx 7.3891$

sum of exponentials ($\sum e^{z_j}$):$$\sum_{j=1}^{3} e^{z_j} = 3.3201 + 1.6487 + 7.3891 = 12.3579$$

normalized attention weights ($\alpha_i$):

$\alpha_1 = \frac{3.3201}{12.3579} \approx \mathbf{0.2687} \quad (26.87\%$

$\alpha_2 = \frac{1.6487}{12.3579} \approx \mathbf{0.1334} \quad (13.34\%)$

$\alpha_3 = \frac{7.3891}{12.3579} \approx \mathbf{0.5979} \quad (59.79\%)$
