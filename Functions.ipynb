**Question 1 : What is the difference between a function and a method in Python?**

**Answer:** The core difference between a function and a method in Python comes down to where they are defined and how they are called.

Ultimately, all methods are functions, but not all functions are methods.

1. Functions are Independent

A function is a free-standing block of code. It doesn't belong to any specific object or data type.

    Where it lives: Defined completely on its own, out in the open.

    How we call it: We use its name directly, followed by parentheses.

    How it gets data: Because it is independent, we have to hand it all the data it needs by passing arguments into those parentheses.

    Example: len(my_list) or print("Hello").

2. Methods are Attached to Objects

A method is essentially a function that is "owned" by a specific object or class. It is a built-in behavior for that specific type of data.

    Where it lives: Defined inside a class, making it dependent on an object.

    How we call it: We must attach it to an existing object using "dot notation" (a period).

    How it gets data: A method automatically has access to the data inside the object it is attached to. We rarely have to pass the object itself into the parentheses.

    Example: my_list.append("apple") or my_string.upper().

**Question 2: Explain the concept of function arguments and parameters in Python**

**Answer:**  A **parameter** is the variable listed inside the parentheses when you define the function. You can think of it as a blank placeholder or an empty box. It tells Python, "When someone uses this function, they need to give me some data to put in here."

An **argument** is the real, actual value you pass into the function when you call it. It is the data that fills the empty box.

For Example: Consider a parameter as a reserved parking space with a sign that says "Employee of the Month". The argument is the actual car that parks in that space on any given day. The space is always there, but the car changes.

**Question 3: What is the purpose of the return statement in a Python function?**

**Answer:** The return statement in Python serves two crucial purposes:

1. it gives data back to whoever called the function, and

2. it immediately exits the function.

Without a return statement, a function does its work in isolation and then quietly vanishes, leaving  with nothing to use later in code. The moment Python hits a return statement, it immediately stops running that function. Any code written below the return statement will be completely ignored.

Because of this, return is often used to stop a function early if a certain condition is met.

````python
'''
Question 4: Write a Python program to calculate the factorial of a number provided by
the user using a while loop
'''
num = int(input("Enter a positive integer: "))

# Initialize variables
factorial = 1
current = num

# Handle invalid, zero, and positive cases
if num < 0:
    print("Factorial does not exist for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1.")
else:
    # Run the while loop until current drops to 0
    while current > 0:
        factorial = factorial * current
        current -= 1  # Decrease current by 1 each iteration

    print(f"The factorial of {num} is {factorial}.")
````

**Question 5: What is Object-Oriented Programming (OOP)?**

**Answer:** Object-Oriented Programming (OOP) is a way of organizing and structuring code by grouping related data and behaviors into individual units called objects.

Instead of writing a long, top-down list of instructions (which is called procedural programming), OOP allows you to model your software after real-world things.

Modern Object-Oriented Programming is built on four fundamental principles that help keep code clean, secure, and reusable:

> Encapsulation: Bundling data and the methods that act on that data into a single unit (the object), and restricting direct access to some of its internal components.

> Abstraction:Hiding the complex, messy implementation details and only showing the essential features to the user.

> Inheritance: Allowing a new class to adopt the properties and behaviors of an existing "parent" class, preventing you from writing the same code twice.

> Polymorphism: Meaning "many forms," this allows different classes to be treated as if they were the same type, often by sharing the same method name but implementing it differently.

**Question 6: What is a class and an object in OOP?**

**Answer:**

1. The Class (The Blueprint)

A class is a set of rules or a template. It defines two things for any object created from it:

    Attributes: The data or properties it will have (e.g., color, size, name).

    Methods: The actions it can perform (e.g., drive, jump, save).

The class itself takes up virtually no memory and doesn't hold actual, specific data. It just says, "If you build one of these, it must have a color, and it must be able to drive."

2. The Object (The Instance)

An object (often called an instance) is the actual, usable thing you build using the class blueprint.

When you create an object, you are filling in the blanks left by the class. You can use one class to stamp out thousands of unique objects, and each object will maintain its own independent state and data.


**Question 7: What is the difference between abstraction and encapsulation?**

**Answer:** Because abstraction and encapsulation both deal with "hiding" things in programming, they are very commonly confused.

> Abstraction hides complexity to make a system easier to use. It focuses on what an object does. Simplify how an object is used by exposing only essential features and hiding the complicated inner workings.

Implemented using abstract base classes (via the abc module) or standard high-level public methods that hide dozens of lower-level internal steps.

> Encapsulation hides internal data to keep a system secure. It focuses on how an object protects its state. Bundle data (attributes) and the methods that operate on that data into a single unit, while restricting direct outside access to keep the data safe from accidental modification.

Implemented using private attributes (prefixed with double underscores like __balance) paired with controlled getter and setter methods to validate changes.



````python
'''
Question 8: Write Python code to create a class Person with attributes name and age,
and a method display that prints these attributes.
'''

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")


# Example usage:
person1 = Person("Alice", 25)
person1.display()
````

**Question 9: What is inheritance in OOP? Provide a simple example in Python.**

**Answer:** Inheritance is a core Object-Oriented Programming mechanism where a child class inherits attributes and methods from an existing parent class. This promotes code reusability and establishes an "is-a" relationship (e.g., a Dog is an Animal).

Example:



```
# Parent Class (Superclass)
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")

# Child Class (Subclass) inheriting from Animal
class Dog(Animal):
    def bark(self):
        print(f"{self.name} says Woof!")

# Create an instance of the child class
dog = Dog("Buddy")

# Access inherited method from the parent
dog.eat()

# Access specific method from the child
dog.bark()
```
Output:
Plaintext

Buddy is eating.
Buddy says Woof!


**Question 10: What is a lambda function in Python? Provide an example**

**Anser:** A lambda function in Python is a small, anonymous function defined without a name using the lambda keyword. Unlike standard functions defined with def, a lambda function is written in a single line and can have any number of arguments, but only one expression (which is automatically evaluated and returned).

> Syntax: lambda arguments: expression

Example:



````python
def square_def(x):
    return x ** 2

# 2. Equivalent lambda function
square = lambda x: x ** 2

print(square(5))
````
