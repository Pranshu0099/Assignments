**Question 1 : What is a RESTful API?**

**Answer:**  A RESTful API (Representational State Transfer) is an architectural style for building web services that allow different applications to communicate over standard HTTP. It treats data entities as resources identified by unique URLs (URIs) and interacts with them using standard HTTP methods.

Core Principles

> Statelessness: Every request from the client contains all the information needed to understand and process it. The server stores no client session context between requests.

> Client-Server Separation: The user interface/client application and the data storage/server backend operate independently and can evolve separately.

> Resource-Based URLs: Endpoints represent nouns (resources), not actions (e.g., /users, /products/42).

**Question 2: What is Flask, and why is it popular for building APIs?**

**Answer:** Flask is a open-source micro-framework for Python designed for web development and building APIs. It is built on top of the Werkzeug WSGI toolkit and the Jinja2 template engine.  Flask is "micro" because it keeps the core framework simple and unopinionated. Unlike full-stack frameworks (like Django), Flask does not enforce a rigid project structure, built-in database ORM, or authentication system.

Reasons for It's Popularity:

1. Minimal Boilerplate: You can create a fully functional, production-ready endpoint in under 10 lines of code.

2. Flexibility and Control: Developers have total freedom over database choices (SQLAlchemy, MongoDB, raw SQL), authentication methods, and project architecture.

3. Extensible Ecosystem: Offers an extensive catalog of extensions (e.g., Flask-RESTful, Flask-CORS, Flask-JWT-Extended, Flask-SQLAlchemy) to add complex features on demand.

4. Easy Routing and JSON Handling: Provides intuitive decorator-based URL routing (@app.route) and built-in JSON serialization via jsonify.

5. Low Learning Curve: Clear, straightforward syntax makes it quick to prototype, test, and deploy microservices.

**Question 3:What are HTTP methods used in RESTful APIs**

**Answer:** HTTP methods (also known as HTTP verbs) define the exact action the client wants the server to perform on a given resource in a RESTful API. They map directly to standard CRUD (Create, Read, Update, Delete) operations.

Primary HTTP Methods

> GET (Read): Retrieves an existing resource or a collection of resources without modifying server state.

> POST (Create): Submits data to create a brand-new subordinate resource on the server.

> PUT (Full Update / Replace): Completely replaces the target resource with the uploaded payload, or creates it if it does not exist.

> PATCH (Partial Update): Applies partial modifications to a resource without replacing the entire entity.

> DELETE (Delete): Removes the specified resource from the server.

**Question 4: What is the purpose of the @app.route() decorator in Flask?**

**Answer:** The @app.route() decorator in Flask is used for URL routing—it binds a specific URL path (endpoint) and optional HTTP methods to a Python function (called a view function).

When an incoming HTTP request hits that URL, Flask automatically executes the associated function and returns its response to the client.

**Usage:**

> URL Mapping: Links a URL pattern directly to business logic without manual routing tables.

> HTTP Method Filtering: Restricts which HTTP methods (GET, POST, PUT, DELETE) the endpoint accepts via the methods argument (defaults to ["GET"]).

> Dynamic Parameter Extraction: Captures dynamic values from the URL (like IDs or usernames) and passes them as arguments to the function.


**Question 5: What is the role of Flask-SQLAlchemy?**

**Answer:**   Flask-SQLAlchemy is an extension for Flask that integrates SQLAlchemy—a powerful Object Relational Mapper (ORM)—into Flask applications.

Its primary role is to simplify database operations by allowing you to interact with SQL databases (like PostgreSQL, MySQL, SQLite) using standard Python objects and classes instead of writing raw SQL queries.

> Object Relational Mapping (ORM): Maps Python classes directly to database tables and class instances to table rows.

> Automated Session & Connection Lifecycle: Automatically manages database connection pools and scoped sessions, tying them cleanly to the Flask request lifecycle (opening connections on request and closing/teardown on response).

> Simplified Configuration: Connects directly to Flask's app.config using standard configuration keys like SQLALCHEMY_DATABASE_URI.

> Database Agnostic: Allows you to switch the underlying database engine (e.g., from SQLite in development to PostgreSQL in production) with minimal or no code changes.

**Question 7:How do you return JSON responses in Flask?**

**Answer:**  Flask provides two primary ways to return JSON responses: using the built-in jsonify() helper function, or by directly returning standard Python dictionaries or lists (supported in modern Flask versions).

1. Using jsonify() (Recommended)

The jsonify() function serializes Python data structures into a JSON-formatted string, wraps it in a Flask Response object, and automatically sets the HTTP header to Content-Type: application/json.

2. Direct Dictionary or List Returns (Flask 1.1+)

If you return a native Python dict or list from a view function, Flask automatically calls jsonify() under the hood.

Why Use jsonify() Over Standard json.dumps()?

> Sets Content-Type Header: json.dumps() only returns a plain string, causing Flask to set the header to Content-Type: text/html. jsonify() explicitly sets application/json.

> Configurable Serialization: jsonify() respects Flask's JSON app configurations (such as key sorting and timestamp formatting).

**Question 8:How do you handle POST requests in Flask?**

**Answer:** To handle POST requests in Flask, you import the global request object, specify methods=["POST"] in the @app.route() decorator, and extract the incoming payload depending on its format (JSON, HTML Form, or URL parameters).

1. Handling JSON Payloads (APIs)

Use request.get_json() to parse incoming JSON payloads into a Python dictionary.

2. Handling HTML Form Data (application/x-www-form-urlencoded)

Use request.form to access fields submitted by standard HTML forms.

3. Handling Both GET and POST in a Single Route

Check request.method to branch your logic between different HTTP methods on the same URL.

**Question 9: How do you handle errors in Flask (e.g., 404)?**

**Answer:** To handle errors in Flask, you use the @app.errorhandler() decorator to register custom error handlers for specific HTTP status codes (like 404 or 500) or specific Python exception classes. You can also trigger errors intentionally using Flask's built-in abort() function.

1. Custom 404 (Not Found) & 500 (Server Error) Handlers

Register handlers to catch HTTP error codes and return consistent JSON responses instead of Flask's default HTML error pages.

2. Manually Triggering Errors with abort()

Use abort() inside a route to halt execution immediately and forward the request to the corresponding error handle.

3. Catching Custom Python Exceptions

You can register @app.errorhandler to catch Python exception classes directly across your whole application

**Question 10: How do you structure a Flask app using Blueprints?**

**Answer:** A Flask Blueprint is an architectural pattern that allows you to organize a web application into independent, self-contained components. In a non-trivial application, defining all routes, models, views, and error handlers inside a single app.py creates tight coupling, circular import hazards, and poor maintainability. Blueprints solve this by acting as recorded sets of operations (routes, templates, static files, middleware) that are later registered onto the central Flask application instance.

1. Scalable Directory Layout

A standard modular Flask architecture uses the Application Factory pattern alongside Blueprints.

2. Decoupling Extensions (app/extensions.py)

To avoid circular imports between models, routes, and the app instance, instantiate extensions without binding them to a specific application immediately.

3. Defining the Blueprints

Each feature folder acts as an isolated module containing its own blueprint definition and routes.

4. The Application Factory (app/__init__.py)

The factory function create_app() dynamically constructs the Flask app, loads environment settings, initializes decoupled extensions, and registers all blueprints with their respective URL prefixes.

5. Execution Script (run.py)

The entry point simply invokes the factory and exposes the WSGI server.
