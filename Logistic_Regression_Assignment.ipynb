**Question 1 : What is Simple Linear Regression (SLR)? Explain its purpose.**

**Answer:** Simple Linear Regression (SLR) is a fundamental statistical method used to model the relationship between a single independent variable (predictor, $x$) and a continuous dependent variable (target, $y$) using a straight line.

Mathematical Equation$$y = \beta_0 + \beta_1 x + \epsilon$$

$y$: Dependent variable (the outcome to predict).

$x$: Independent variable (the input feature).

$\beta_0$: $y$-intercept (the expected value of $y$ when $x = 0$).

$\beta_1$: Slope coefficient (the change in $y$ for a 1-unit increase in $x$).

$\epsilon$: Random error term (unexplained variance).

**Core PurposePrediction & Forecasting:**

 1. Predict unknown future values of a target variable based on a given input value (e.g., predicting salary based on years of experience).

 2. Quantifying Relationships: Determine the direction (positive or negative) and magnitude of the impact that an independent variable has on a outcome variable.

 3. Baseline Benchmarking: Serve as a simple, interpretable benchmark model before attempting more complex non-linear or multi-variable machine learning algorithms.

**Question 2: What are the key assumptions of Simple Linear Regression?**

**Answer:** Simple Linear Regression models the relationship between a predictor variable $X$ and a target variable $Y$ using the equation $Y = \beta_0 + \beta_1 X + \epsilon$. For Ordinary Least Squares (OLS) estimates to be unbiased and for statistical tests to be valid, four core assumptions—summarized by the acronym LINE—must hold.

1. Linearity: The relationship between $X$ and the mean of $Y$ is linear. Fitting a straight line to non-linear data produces biased parameter estimates.

2. Independence: The residuals (error terms $\epsilon_i$) are independent of one another. There is no autocorrelation between consecutive residuals, which commonly occurs in time-series data.

3. Normality: The residuals are normally distributed with a mean of zero ($\epsilon \sim N(0, \sigma^2)$). This assumption is essential for accurate $p$-values, $t$-tests, $F$-tests, and confidence intervals.

4. Equal Variance (Homoscedasticity): The variance of the error terms remains constant across all levels of $X$.

Additional Considerations

**Measured Without Error:** The independent variable $X$ is assumed to be measured precisely without random noise.

**Sensitivity to Outliers:** Extreme values or high-leverage points can disproportionately pull the regression line and distort slope coefficients.

**Question 5: What is the method of least squares in linear regression?**

**Answer:** The Method of Least Squares (or Ordinary Least Squares, OLS) is an optimization technique used in linear regression to find the line of best fit for a dataset.

It determines the line's parameters ($\hat{\beta}_0$ and $\hat{\beta}_1$) by minimizing the sum of the squared vertical distances (residuals) between the observed data points and the regression line.

**Interpretation**

For each observation $i$, the residual $e_i$ represents the prediction error:$$e_i = Y_i - \hat{Y}_i = Y_i - (\hat{\beta}_0 + \hat{\beta}_1 X_i)$$The Method of Least Squares minimizes the Sum of Squared Residuals (SSR):$$\text{SSR} = \sum_{i=1}^{n} e_i^2 = \sum_{i=1}^{n} \left(Y_i - (\hat{\beta}_0 + \hat{\beta}_1 X_i)\right)^2$$

**Why Residuals Are Squared**

1. Eliminates Cancellation: Unsquared residuals sum to zero ($\sum e_i = 0$) because positive and negative errors offset each other. Squaring ensures all errors are positive.

2. Penalizes Outliers: Squaring gives larger weight to points far from the line (e.g., an error of 4 contributes 16 to SSR, whereas an error of 2 contributes only 4).

3. Solvable via Calculus: The resulting SSR function is smooth and convex, allowing exact closed-form solution formulas via partial derivatives.

**Question 6: What is Logistic Regression? How does it differ from Linear Regression?**

**Answer:** Logistic Regression is a statistical model used to predict the probability of a categorical outcome (most commonly binary, such as Yes/No, Pass/Fail, or Spam/Not Spam) based on one or more predictor variables.

Instead of fitting a straight line that can output any real value, it transforms a linear equation using the Sigmoid (logistic) function to constrain predictions strictly between 0 and 1:$$P(Y = 1 \mid X) = \sigma(\beta_0 + \beta_1 X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 X)}}$$

**Key Differences: Linear vs. Logistic Regression**

**Target Output and Task Type**

Linear Regression predicts continuous numerical values (e.g., home prices, temperatures, or revenues). Logistic Regression predicts categorical class membership or event probabilities (e.g., spam vs. not spam, pass vs. fail, or disease presence).

**Output Range**

Linear Regression predictions are unbounded and can theoretically range from negative infinity to positive infinity ($-\infty, +\infty$). Logistic Regression output is strictly bounded between $0$ and $1$ because it outputs a probability.

**Function Shape**

Linear Regression models relationships using a straight line ($\hat{Y} = \beta_0 + \beta_1 X$). Logistic Regression transforms a linear equation into an S-shaped Sigmoid curve ($p = \frac{1}{1 + e^{-z}}$) to account for non-linear probability transitions.

**Parameter Fitting Method**

Linear Regression uses Ordinary Least Squares (OLS) to directly minimize the sum of squared errors between actual and predicted points. Logistic Regression uses Maximum Likelihood Estimation (MLE), an iterative optimization process that finds coefficient values maximizing the likelihood of observing the actual binary outcomes.

**Residual and Variance Assumptions**

Linear Regression relies on key residual assumptions: errors must be normally distributed with constant variance (homoscedasticity). Logistic Regression does not assume normally distributed or homoscedastic errors, as binary outcome errors naturally follow a Bernoulli distribution.

**Decision Logic**

Linear Regression returns the raw predicted numeric value directly. Logistic Regression converts calculated probabilities into discrete predictions by applying a cutoff threshold (for example, classifying an outcome as positive if $p \ge 0.5$).

**Question 7: Name and briefly describe three common evaluation metrics for regression models.**

**Answer:** Three Primarily used Regression model evaluation mechanisms are:

**1. Mean Absolute Error (MAE)** $$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \vert{}Y_i - \hat{Y}_i\vert{}$$Measures the average magnitude of absolute differences between actual values ($Y_i$) and predicted values ($\hat{Y}_i$). It is easy to interpret because it is in the same units as the target variable, and it treats all individual errors linearly without overly penalizing outliers.

**2. Root Mean Squared Error (RMSE)**  $$\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2}$$Measures the square root of the average squared differences between actual and predicted values. Because errors are squared before being averaged, RMSE penalizes larger errors more heavily than smaller ones, making it ideal when large prediction mistakes are especially costly.

**3. Coefficient of Determination ($R^2$)** $$R^2 = 1 - \frac{\sum (Y_i - \hat{Y}_i)^2}{\sum (Y_i - \bar{Y})^2}$$Measures the proportion of variance in the dependent variable that is predictable from the independent variable. It ranges from $0$ to $1$ (or $0\%$ to $100\%$), where $1$ indicates that the model accounts for all variability in the target data relative to the baseline mean ($\bar{Y}$).

**Question 8: What is the purpose of the R-squared metric in regression analysis?**

**Answer:** The primary purpose of $R^2$ (Coefficient of Determination) is to measure the goodness-of-fit of a regression model by quantifying the proportion of variance in the dependent variable ($Y$) that is explained by the independent variable(s) ($X$).

**Core Purposes in Regression Analysis**

**Quantifying Predictive Power:** $R^2$ indicates how much better a model predicts outcomes compared to simply using the sample mean ($\bar{Y}$) as a baseline prediction.

**Standardized Metric:** Because $R^2$ ranges from $0$ to $1$ ($0\%$ to $100\%$), it provides a scale-independent score that allows direct comparison across models trained on different units or scales.

**Evaluating Variable Contribution:** In linear models, $R^2$ reflects the collective explanatory strength of the included features.

*Limitations*

1. Does Not Guarantee Correct Model Shape: A high $R^2$ does not mean the model is linear; non-linear patterns can still produce high $R^2$ values if overall dispersion is low.

2. Always Increases with Added Features: Adding more predictor variables—even irrelevant noise—never decreases $R^2$. To account for feature count, Adjusted $R^2$ is used instead.

```python
"""
Question 9: Write Python code to fit a simple linear regression model using
scikit-learn and print the slope and intercept.
"""


import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([[1], [2], [3], [4], [5]])  # Features (must be 2D)
y = np.array([2, 4, 5, 4, 5])            # Target values

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Print the slope (coefficient) and intercept
print(f"Slope (coefficient): {model.coef_[0]:.4f}")
print(f"Intercept: {model.intercept_:.4f}")

# Generate predictions for the regression line
X_line = np.linspace(0, 6, 100).reshape(-1, 1)
y_line = model.predict(X_line)

# Plot
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', s=100, label='Data Points', zorder=5)
plt.plot(X_line, y_line, color='red', linewidth=2, label='Regression Line')
plt.xlabel('X', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Simple Linear Regression', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

```

**Question 10: How do you interpret the coefficients in a simple linear regression model?**

**Answer:** In the simple linear regression model equation $\hat{Y} = \hat{\beta}_0 + \hat{\beta}_1 X$, the parameters $\hat{\beta}_0$ and $\hat{\beta}_1$ describe the baseline starting value and the expected rate of change between the variables.

**Intercept** ($\hat{\beta}_0$)

**Interpretation:** Represents the expected average value of the dependent variable ($Y$) when the independent variable ($X$) is equal to zero ($X = 0$).

It has meaningful real-world interpretation only if $X = 0$ is conceptually plausible and lies within the range of observed data (e.g., zero ad spend). If $X = 0$ is impossible (e.g., a house with zero square feet), the intercept serves purely as a mathematical anchor to position the regression line.

**Slope Coefficient** ($\hat{\beta}_1$)

**Interpretation:** Represents the estimated average change in $Y$ associated with a one-unit increase in $X$.

**Positive Value** ($\hat{\beta}_1 > 0$): Indicates a positive relationship; for every 1-unit increase in $X$, $Y$ increases on average by $\hat{\beta}_1$ units.

**Negative Value** ($\hat{\beta}_1 < 0$): Indicates an inverse relationship; for every 1-unit increase in $X$, $Y$ decreases on average by $\vert{}\hat{\beta}_1\vert{}$ units.

**Zero Value** ($\hat{\beta}_1 = 0$): Indicates no linear relationship between $X$ and $Y$; changes in $X$ do not alter the expected value of $Y$.
