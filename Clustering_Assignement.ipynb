# Theoretical Questions

**1. What is unsupervised learning in the context of machine learning?**

**Answer:** Unsupervised learning is a ML paradigm where models learn underlying patterns, groupings, or structures directly from unlabeled input data without explicit target output labels or external supervision. Common tasks include clustering, dimensionality reduction, and association rule learning.

---

**2. How does K-Means clustering algorithm work?**

**Answer:** K-Means partitions data into $K$ distinct clusters through an iterative process:

> Initialization: Select $K$ initial cluster centroids (at random or via K-Means++).

> Assignment: Assign each data point to its nearest centroid based on a distance metric (typically Euclidean distance).

> Update: Recalculate each centroid as the mean of all data points assigned to that cluster.Convergence: Repeat the assignment and update steps until centroids stabilize or max iterations are reached.

---

**3. Explain the concept of a dendrogram in hierarchical clustering.**

**Answer:** A dendrogram is a tree-like visual representation of the sequential merging or splitting of clusters in hierarchical clustering.

> Structure: The horizontal axis displays data points or sub-clusters, while the vertical axis indicates the distance or dissimilarity threshold at which merges occur.

> Cluster Determination: Cutting the dendrogram horizontally at a chosen height defines the final number of clusters ($K$).

---

**4. What is the main difference between K-Means and Hierarchical Clustering?**

**Answer:**

Cluster Specification: K-Means requires specifying $K$ upfront, whereas Hierarchical Clustering creates a continuous multi-level tree without requiring prior $K$.

Computational Scalability: K-Means is computationally fast $\mathcal{O}(n)$, making it suitable for large datasets. Hierarchical Clustering is slow $\mathcal{O}(n^3)$, making it limited to smaller datasets.

Determinism: K-Means results can vary due to random centroid initialization, while Hierarchical Clustering yields deterministic results for a given linkage metric.

---

**5. What are the advantages of DBSCAN over K-Means?**

**Answer:**

> Arbitrary Cluster Shapes: DBSCAN can identify arbitrarily shaped clusters (e.g., non-spherical, nested rings), whereas K-Means assumes spherical clusters.

> No Pre-specified K: DBSCAN automatically determines the cluster count based on local point density.

> Noise Handling: DBSCAN explicitly isolates noise points and outliers instead of forcing them into clusters.

---

**6. When would you use Silhouette Score in clustering?**

**Answer:**
 Silhouette Score is used to evaluate cluster separation and compactness when ground-truth class labels are unavailable. It measures how close each point is to its own cluster compared to neighboring clusters, aiding in selecting the optimal number of clusters ($K$).

---

 **7. What are the limitations of Hierarchical Clustering?**

 **Answer:**

 > High Time and Memory Complexity: Requires $\mathcal{O}(n^3)$ execution time and $\mathcal{O}(n^2)$ memory, scaling poorly on large datasets.

 > Irreversible Merges: Decisions made in earlier merging steps cannot be undone later in the pipeline.


 > Sensitivity to Noise: Outliers significantly impact distance calculations and distort the hierarchical tree.

---

 **8. Why is feature scaling important in clustering algorithms like K-Means?**

**Answer:**
Distance-based algorithms rely on metric calculations like Euclidean distance. Features with larger numerical ranges (e.g., income in thousands) dominate distance values over features with smaller ranges (e.g., age in tens). Feature scaling (e.g., StandardScaler) ensures all attributes contribute equally to distance calculations.

---

**9. How does DBSCAN identify noise points?**

**Answer:** A point is labeled as noise if:

It contains fewer than minPts within its $\epsilon$-neighborhood (i.e., it is not a Core Point).

It does not fall within the $\epsilon$-neighborhood of any Core Point (i.e., it is not a Border Point).

---

**10. Define inertia in the context of K-Means.**

**Answer:** Inertia (Within-Cluster Sum of Squares, WCSS) measures internal cluster tightness. It is the sum of squared Euclidean distances from each sample $x_i$ to its assigned cluster centroid $\mu_k$:$$\text{Inertia} = \sum_{k=1}^{K} \sum_{x_i \in C_k} \Vert{}x_i - \mu_k\Vert{}^2$$

---

**11. What is the elbow method in K-Means clustering?**

**Answer:** It is a heuristic technique for identifying optimal $K$:

Process: Plot Inertia (WCSS) against a range of candidate $K$ values.Interpretation: As $K$ increases, inertia drops. The point where the rate of decrease sharply bends (the "elbow") represents the point of diminishing returns and the optimal $K$.

---

**12. Describe the concept of "density" in DBSCAN.**

**Answer:** Density in DBSCAN is defined by the number of data points residing within a spatial radius $\epsilon$ (Epsilon) of a given target point. A region is considered dense if it contains at least minPts within distance $\epsilon$.

---

**13. Can hierarchical clustering be used on categorical data?**

**Answer:** Yes. While standard Euclidean distance requires continuous data, hierarchical clustering can operate on categorical features by using distance metrics designed for non-numeric data, such as Gower's distance, Jaccard dissimilarity, or Hamming distance.

---

**14. What does a negative Silhouette Score indicate?**

**Answer:** A negative score (nearing $-1$) indicates that a data point has been assigned to the wrong cluster, as it is on average closer to points in a neighboring cluster than to points in its assigned cluster.

---

**15. Explain the term "linkage criteria" in hierarchical clustering.**

**Answer:** Linkage criteria determine how the distance between two clusters is measured during hierarchical merging:

1. Single Linkage: Distance between the two closest points across clusters.
2. Complete Linkage: Distance between the two farthest points across clusters.
3. Average Linkage: Average distance between all point pairs across clusters.
4. Ward's Linkage: Minimizes the increase in within-cluster variance after merging.

---

**16. Why might K-Means clustering perform poorly on data with varying cluster sizes or densities?**

**Answer:** K-Means naturally minimizes total variance under the assumption of equal-sized spherical clusters.Varying Sizes: Larger clusters get split artificially into smaller components.Varying Densities: Centroids pull toward dense regions, misclassifying adjacent sparse clusters.

---

**17. What are the core parameters in DBSCAN, and how do they influence clustering?**

**Answer:** The main paramenters include:

> $\epsilon$ (Epsilon): The maximum neighborhood search radius. Setting it too small leaves points as noise; setting it too large merges distinct clusters.

> minPts: The minimum number of points inside an $\epsilon$-neighborhood required to form a Core Point. Higher values require higher density and increase noise tolerance.

---

**18. How does K-Means++ improve upon standard K-Means initialization?**

**Answer:** Standard K-Means picks initial centroids purely at random, risking poor local minima. K-Means++ picks the first centroid randomly, then chooses remaining centroids with a probability proportional to their squared distance from existing centroids ($P(x) \propto D(x)^2$). This ensures centroids are spread out across the feature space, speeding up convergence and improving overall clustering quality.

---

**19. What is agglomerative clustering?**

**Answer:** Agglomerative clustering is a bottom-up hierarchical approach. It starts with every individual point as its own cluster, then iteratively merges the closest pair of clusters based on a specified distance metric and linkage criterion until a single cluster (or target count) remains.

---

**20. What makes Silhouette Score a better metric than just inertia for model evaluation?**

**Answer:**

> Inertia: Only evaluates cohesion (internal compactness) and monotonically decreases toward zero as $K$ increases, biasing evaluation toward higher $K$.

> Silhouette Score: Balances both cohesion (distance within cluster) and separation (distance to nearest neighboring cluster), bounded between $-1$ and $+1$, allowing objective comparison across different values of $K$.

```python
# 29. Create synthetic data using make_blobs and evaluate silhouette scores for $k = 2$ to $5$. Display as a bar chart.

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score

# Generate synthetic data
X, _ = make_blobs(n_samples=300, centers=4, random_state=42)

# Evaluate Silhouette Scores
k_values = [2, 3, 4, 5]
scores = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    scores.append(silhouette_score(X, labels))

# Plot bar chart
plt.bar(k_values, scores, color='skyblue', edgecolor='black')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs k")
plt.show()
```

```python
# 30. Load the Iris dataset and use hierarchical clustering to group data. Plot a dendrogram with average linkage.

import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
X = iris.data

# Compute average linkage matrix
linked = linkage(X, method='average')

# Plot dendrogram
plt.figure(figsize=(10, 5))
dendrogram(linked, truncate_mode='lastp', p=30)
plt.title("Iris Dataset Dendrogram (Average Linkage)")
plt.xlabel("Sample Index or Cluster Size")
plt.ylabel("Distance")
plt.show()
```

```python
# 31. Generate synthetic data with overlapping clusters using make_blobs, then apply K-Means and visualize with decision boundaries.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Overlapping blobs
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=2.5, random_state=42)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X)

# Meshgrid for decision boundaries
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))
Z = kmeans.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

# Visualize decision boundary
plt.contourf(xx, yy, Z, alpha=0.3, cmap='tab10')
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='tab10', edgecolors='k')
plt.title("K-Means Decision Boundaries on Overlapping Blobs")
plt.show()
```

```python
# 34. Load the Breast Cancer dataset and compare inertia values for $K = 2$ to $6$ using K-Means. Show results in a line plot.

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_breast_cancer

# Load dataset
cancer = load_breast_cancer()
X = cancer.data

# Calculate inertia values for K=2 to 6
k_range = range(2, 7)
inertias = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
    inertias.append(kmeans.inertia_)

# Line plot
plt.plot(k_range, inertias, marker='o', linestyle='--', color='b')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Plot (Inertia vs K) on Breast Cancer Dataset")
plt.show()
```

```python
# 44. Load the Iris dataset and plot the Silhouette Coefficient for each sample after KMeans clustering.

import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.metrics import silhouette_samples, silhouette_score

# Load Iris and run KMeans
iris = load_iris()
X = iris.data
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Calculate sample-wise and overall silhouette values
sample_silhouette_values = silhouette_samples(X, labels)
avg_score = silhouette_score(X, labels)

plt.figure(figsize=(8, 6))
y_lower = 10
for i in range(3):
    ith_cluster_silhouette_values = sample_silhouette_values[labels == i]
    ith_cluster_silhouette_values.sort()
    size_cluster_i = ith_cluster_silhouette_values.shape[0]
    y_upper = y_lower + size_cluster_i

    plt.fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        ith_cluster_silhouette_values,
        alpha=0.7,
    )
    plt.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))
    y_lower = y_upper + 10

plt.axvline(
    x=avg_score,
    color="red",
    linestyle="--",
    label=f"Average Score ({avg_score:.2f})",
)
plt.xlabel("Silhouette Coefficient Values")
plt.ylabel("Cluster Label")
plt.title("Sample-wise Silhouette Coefficients (Iris Dataset)")
plt.legend()
plt.show()
```

```python
# 46. Load the Wine dataset, apply KMeans, and visualize the cluster assignments in a seaborn pairplot (first 4 features).

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.datasets import load_wine

# Wine data (first 4 features)
wine = load_wine()
df = pd.DataFrame(wine.data, columns=wine.feature_names).iloc[:, :4]

#  KMeans predictions
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(df).astype(str)

#  Pairplot
sns.pairplot(df, hue='Cluster', palette='tab10')
plt.suptitle(
    "Wine Dataset - First 4 Features Pairplot with KMeans Clusters", y=1.02
)
plt.show()
```

```python
# 47. Generate noisy blobs using make_blobs and use DBSCAN to identify both clusters and noise points. Print the count.

import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs

# Generate noisy blobs
X, _ = make_blobs(n_samples=400, centers=3, cluster_std=2.0, random_state=42)

# Fit DBSCAN
dbscan = DBSCAN(eps=1.0, min_samples=5)
labels = dbscan.fit_predict(X)

# Calculate counts
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = np.sum(labels == -1)

print(f"Number of clusters identified: {n_clusters}")
print(f"Number of noise points identified: {n_noise}")
```

```python
# 48. Load the Digits dataset, reduce dimensions using t-SNE, then apply Agglomerative Clustering and plot the clusters.

import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE

# Load Digits dataset and reduce using t-SNE
digits = load_digits()
X_tsne = TSNE(n_components=2, random_state=42).fit_transform(digits.data)

# Perform Agglomerative Clustering
agg = AgglomerativeClustering(n_clusters=10)
labels = agg.fit_predict(X_tsne)

# Plot 2D representation
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels, cmap='tab10', alpha=0.7)
plt.title("Agglomerative Clustering on t-SNE Reduced Digits Dataset")
plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")
plt.show()
```
