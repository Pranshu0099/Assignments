**Question 1 : What is mutability in Python?**

**Answer:**  Mutability refers to whether an object's state or contents can be changed after it has been created in memory. Every variable in Python points to an object in memory, and every object is strictly one of these two types:

1. Mutable: We can add, remove, or alter items inside the object without changing its memory address. For example: list, dict, set, most user-defined classes.

2. Immutable: The object is locked down. "Modifying" it actually creates a brand-new object in a different memory location. Example: int, float, bool, str (string), tuple, frozenset

> The easiest way to prove mutability is by using Python's built-in id() function, which returns the exact memory address of an object.

> Understanding this concept is crucial because it affects how variables behave when you pass them around your code.

**Question 3:What are logical operators in Python?**

**Answer:** Logical operators in Python are used to combine multiple conditions or reverse the boolean value (True or False) of a statement.

Python keeps this simple by using plain English words rather than symbols (like && or || used in other languages). There are only three logical operators:

> and	-  Returns True only if both statements are true. - 	x > 5 and x < 10

> or	-  Returns True if at least one statement is true. - x == 5 or x == 10

> not	-  Reverses the result (turns True to False, and vice versa).	- not(x > 5)




````python
# **Question 4: Write a Python program to convert user input from string to integer, float, and boolean types.**

user_input = input("Enter a number: ")

#  Convert the string to different types
my_integer = int(user_input)
my_float = float(user_input)
my_boolean = bool(user_input)


print("As a string:", user_input)
print("As an integer:", my_integer)
print("As a float:", my_float)
print("As a boolean:", my_boolean)
````

**Question 5: Describe how dictionaries store data.**

**Answer:** Python dictionaries store data as key-value pairs, rather than in a numbered sequence like lists or tuples.

Rules Of Dictionaries:

> Uniqueness: Must be strictly unique. You cannot have two identical keys.	Can be duplicated. Multiple keys can hold the exact same value.

> Data Types: Must be immutable (strings, numbers, or tuples). We cannot use a list as a key.	Can be anything (lists, strings, numbers, or even other dictionaries).

When we give a dictionary a key, Python uses a mathematical process called hashing. It runs key through a formula that instantly calculates the exact memory address where the value is stored.

This is why looking up data in a dictionary is incredibly fast. Even if a dictionary contains millions of items, Python doesn't have to scan through it from beginning to end; it simply calculates the key's "hash" and jumps straight to the answer.

**Question 7: What is the difference between for and while loops?**

**Answer:** When I write Python, the main difference for me comes down to whether I know exactly how many times I need to loop before I start.

For loops are for when you have a specific collection of items to go through, like a list, or when you know exactly how many times you want the loop to run.

While loops are for when you don't know how many times I need to loop. I just want the code to keep running until a specific condition changes.

The main practical difference is that a for loop stops on its own when it reaches the end of the items. With a while loop, we have to manually change something inside the loop to make the condition false, or it will just keep running forever.

````python
#5Question 8: Write a Python program to calculate the factorial of a number provided by the user using a while loop.

num = int(input("Enter a positive integer: "))

334
if num < 0:
    print("Factorial does not exist for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1.")
else:
    factorial = 1
    current = num

    # Run the while loop until current drops to 0
    while current > 0:
        factorial = factorial * current
        current -= 1  # Decrease current by 1 each time

    print(f"The factorial of {num} is {factorial}.")
````

**Question 9: Why might you use a set instead of a list in Python?**

**Answer:**  To enforce uniqueness or perform lightning-fast lookups.

If the order of data matters, or if you need to keep duplicate items, we must use a list. But if those two things don't matter, a set is often the better tool.

***Three main reasons to choose a set:***

1. A set mathematically cannot contain duplicate values. If you try to add a duplicate, Python just ignores it. This makes sets the perfect tool for cleaning up data.

2. to check if a specific item exists inside a massive collection of items, a set is drastically faster than a list.

3. to use mathematical operations to compare two groups of data instantly, which would require messy for loops if you used lists.
