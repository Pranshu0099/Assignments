**Question 1**: What is the difference between descriptive statistics and inferential statistics? Explain with examples.
-
**Answer**: Statistics deals with collecting, organizing, analyzing, interpreting, and presenting data. Differential and inferential are two broad classifications of Statistics.

1. **Descriptive statistics** includes the methods used to summarize and present data through numerical measures and graphical representations strictly based on available data. Measures of Central Tendency and spread variance are the primary methods used based on the requirements.

2. **Inferential statistics** uses data sampling to make conclusions or predictions about an entire data population when the dataset is very large. , Inferential statistics helps estimate population characteristics based on a representative sample. Hypothesis testing and Regression analysis are common methods of this type.

Descriptive statistics helps organize and summarize collected data, whereas inferential statistics uses sample data to make predictions or draw conclusions about a larger population. Both are essential in data science because descriptive statistics helps understand existing data, while inferential statistics supports evidence-based decision making and forecasting.

**Example**: An e-commerce company records the daily orders received in one month.

1. Using descriptive statistics, the analyst calculates the average daily orders, maximum orders, minimum orders, and creates graphs showing sales trends.
2. Using inferential statistics, the analyst predicts future monthly sales based on a sample of previous customer transactions and determines whether a new marketing campaign has significantly increased sales.





**Question 2:** What is sampling in statistics? Explain the differences between random and stratified sampling.
-

**Answer:** Sampling refers to selection of a subset of a larger population to procees and generate the insights used for decision making. It is preferred to make cost effective and time saving business decisions. Primary Sampling methods in common use are:

1. Random Sampling: We pick the samples at a random frequency and gap giving a equal chance to each individual unit from the population to get selected.
2. Stratified Sampling: In this method, we divide the population in groups termed as strata based on common characterstics. Then a random Sample is taken from each individual strata for further processing.

**Difference between Random and Stratified Sampling:**

1. Selection is random in random sampling whereas selection is further done from homogenous groups called strata.

2. Random sampling may miss out on any sample sets with less number of representative individuals whereas being grouped in strata gives an advantage to each group for atleaset 1 sample to be selected in stratified sampling.

3. Random Sampling is better when the dataset is comparatively smaller and homogenous but Stratified sampling should be preferred if the population is large.

4. Stratified sampling is more accurate in terms of observations as compared to random when the population is larger.

**Question 4:** Explain skewness and kurtosis. What does a positive skew imply about the data?

Answer: Skewness and kurtosis are statistical measures used to describe the shape of a data distribution. They help analysts understand whether data is symmetrical, how values are spread, and whether extreme values are present.


##  Skewness
*Skewness* measures the degree of asymmetry in a distribution around its mean value.

**Types of Skewness**

(a) Zero Skewness

* Distribution is perfectly symmetrical.
* Mean = Median = Mode.
* Example: Ideal normal distribution.

(b) Positive Skewness (Right-Skewed)

* The tail extends toward the right.
* A few unusually large values pull the mean upward.
* Mean > Median > Mode.

*Example:*
Income distribution in a country where most people earn moderate incomes, while a few individuals earn extremely high salaries.


(c) Negative Skewness (Left-Skewed)

* The tail extends toward the left.
* A few unusually small values pull the mean downward.
* Mean < Median < Mode.

*Example:*
Marks in an easy examination where most students score high but a few score very low.

##  Kurtosis

*Kurtosis* measures the heaviness of the tails of a distribution and indicates the likelihood of extreme values (outliers).

**Types of Kurtosis**

(a) Mesokurtic

* Similar to a normal distribution.
* Moderate peak and moderate tails.
* Kurtosis ≈ 3.

(b) Leptokurtic

* Tall, sharp peak.
* Heavy tails.
* Higher probability of outliers.

*Example:* Daily stock market returns.

(c) Platykurtic

* Flatter peak.
* Light tails.
* Lower probability of extreme values.

*Example:* Uniformly distributed data.


*Importance of Skewness and Kurtosis*

* Help understand the shape of data.
* Detect outliers and unusual observations.
* Assist in selecting suitable statistical methods.
* Improve predictive models in data science.
* Help verify assumptions required for hypothesis testing and regression analysis.


## What Does a Positive Skew Imply about the data?

A positive skew indicates that:

* Most observations are concentrated at lower values.
* A small number of unusually high values create a long right tail.
* The mean becomes greater than the median because it is influenced by high outliers.
* The distribution is not symmetric.

*Example*

Customer purchase amounts in an online store often show positive skewness. Most customers spend moderate amounts, while a few customers place very large orders, increasing the average purchase value.

```python
"""
Question 5: Implement a Python program to compute the mean, median, and mode of
a given list of numbers.

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
"""


from statistics import mean, median, multimode

# For any n number of user inputs
user_input = input("Enter the numbers: ")
numbers = [int (n) if n.is_integer() else n for n in
          [float(x) for x in user_input.replace(","," ").split()]]

# Calculate statistics
mean_value = mean(numbers)
median_value = median(numbers)
mode_value = multimode(numbers)

# Display results
print("Numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

```python
"""
Question 6: Compute the covariance and correlation coefficient between the
following two datasets provided as lists in Python.

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
"""
#Answer is given below as python code

import numpy as np

''' Given datasets - Taking the dataset as given in question instead of
asking for dynamic user input as previous Question'''

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Calculate covariance matrix
covariance_matrix = np.cov(list_x, list_y)

# Calculate correlation coefficient
correlation = np.corrcoef(list_x, list_y)

# Print results
print("Covariance Matrix:")
print(covariance_matrix)

print("\nCorrelation Coefficient Matrix:")
print(correlation)

print("\nCovariance:", covariance_matrix[0,1])
print("Correlation Coefficient:", correlation[0,1])
```

```python
"""
Question 7: Write a Python script to draw a boxplot for the following
numeric list and identify its outliers. Explain the result:

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
"""
#Answer is given below as python code

import matplotlib.pyplot as plt
import numpy as np

# Given data
user_input = input("Enter the numbers: ")
data = [int (n) if n.is_integer() else n for n in
          [float(x) for x in user_input.replace(","," ").split()]]

# Draw boxplot
plt.figure(figsize=(8,5))
plt.boxplot(data, vert=True, patch_artist=True)
plt.title("Box Plot of Given Data")
plt.ylabel("Values")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()

# Calculate Quartiles
Q1 = np.percentile(data, 25)
Q2 = np.median(data)
Q3 = np.percentile(data, 75)

# Interquartile Range
IQR = Q3 - Q1

# Outlier Limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find Outliers
outliers = [x for x in data if x < lower_limit or x > upper_limit]

print("Q1 =", Q1)
print("Median =", Q2)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Limit =", lower_limit)
print("Upper Limit =", upper_limit)
print("Outliers =", outliers)
```

## Explanation of Q. 7 Code

A *box plot* is a graphical representation that summarizes a dataset to understand the distribution, data spread, median position and any outliers in the given data using the five-number summary:

* Minimum
* First Quartile (Q1)
* Median (Q2)
* Third Quartile (Q3)
* Maximum


**Step 1:** Quartiles

For the given data:

* *Q1 = 15.75*
* *Median (Q2) = 21*

**Step 2:** Calculate the Interquartile Range (IQR)

The Interquartile Range measures the spread of the middle 50% of the data.

*IQR = Q3 − Q1 = 23.25 − 15.75 = 7.5*

**Step 3:** Determine Outlier Limits

Lower Limit:

*Q1 − 1.5 × IQR = 15.75 − 11.25 = 4.5*

Upper Limit:

*Q3 + 1.5 × IQR = 23.25 + 11.25 = 34.5*

Any value below *4.5* or above *34.5* is considered an outlier.

**Step 4:** Identify Outliers

Among the given values, only *35* is greater than *34.5*.

Therefore,  *Outlier = 35*

---

## Interpretation

The box plot shows that:

* Most observations lie between *14 and 29*.
* The *median is 21*, indicating that half the observations are below 21 and half are above it.
* The middle 50% of the observations lie between *Q1 (15.75)* and *Q3 (23.25)*.
* The value *35* is unusually high compared to the rest of the dataset and is therefore identified as an *outlier*.

Since there is only one high outlier, the distribution exhibits a *slight positive (right) skew*.

The above code successfully created a box plot and identified outliers using the *1.5 × IQR rule. For the given dataset, **35* is the only outlier.

##**Question 8:**
You are working as a data analyst in an e-commerce company. The
marketing team wants to know if there is a relationship between advertising spend and daily sales.

● Explain how you would use covariance and correlation to explore this
relationship.

● Write Python code to compute the correlation between the two lists:

advertising_spend = [200, 250, 300, 400, 500]

daily_sales = [2200, 2450, 2750, 3200, 4000]


###**Answer:**  

As a data analyst, I would first collect data on *advertising expenditure* and the corresponding *daily sales* over a period of time.

To determine whether advertising affects sales, I would use *covariance* and *correlation*.

## Covariance

Covariance indicates whether two variables move in the same direction or in opposite directions.

* *Positive covariance:* Both variables increase together.
* *Negative covariance:* One variable increases while the other decreases.
* *Zero covariance:* No clear linear relationship.

However, covariance does not indicate the *strength* of the relationship because its value depends on the units of measurement.

## Correlation

Correlation measures both the *strength* and *direction* of the linear relationship between two variables.

The correlation coefficient ranges from *−1 to +1*.

* *+1* → Perfect positive relationship
* *0* → No linear relationship
* *−1* → Perfect negative relationship

A high positive correlation suggests that higher advertising spending is associated with higher daily sales.


Although covariance tells us the direction of the relationship, correlation is preferred because:

* It is standardized.
* It is independent of measurement units.
* It is easier to compare across different datasets.
* It clearly indicates the strength of the relationship.

```python
"""
Python code to compute the correlation between the two lists:

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

"""

import numpy as np

# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Correlation matrix
correlation_matrix = np.corrcoef(advertising_spend, daily_sales)

# Correlation coefficient
correlation = correlation_matrix[0, 1]

print("Correlation Matrix:")
print(correlation_matrix)

print("\nCorrelation Coefficient:", correlation)
```

###Question 9:
Your team has collected customer satisfaction survey data on a scale of
1-10 and wants to understand its distribution before launching a new product.


● Explain which summary statistics and visualizations (e.g. mean, standard
deviation, histogram) you’d use.

● Write Python code to create a histogram using Matplotlib for the survey data:

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

### Answer:

(a) Summary Statistics and Visualizations

Before launching a new product, it is important to analyze customer satisfaction scores to understand how customers perceive existing products. Summary statistics and graphical visualizations provide valuable insights into customer opinions.

1. Mean (Average)

The mean represents the average satisfaction score.

For the given dataset:

* Total Score = *110*
* Number of Responses = *15*

*Mean = 110 ÷ 15 = 7.33*

The average customer satisfaction is approximately *7.33 out of 10*, indicating that customers are generally satisfied.


2. Median

The median is the middle value after arranging the scores in ascending order.

Sorted data:

*4, 5, 6, 6, 7, 7, 7, 7, 8, 8, 8, 9, 9, 9, 10*

Median = *7*

This shows that half of the customers rated the product *7 or below, while the remaining half rated it **7 or above*.


3. Mode

The mode is the most frequently occurring value.

In the given dataset:

* *7* appears *4 times*, more than any other score.

Mode = *7*

This indicates that the most common customer rating is *7*.

4. Standard Deviation

Standard deviation measures how much the ratings vary from the average.

* A *small standard deviation* indicates that customer opinions are consistent.
* A *large standard deviation* indicates greater variation in customer opinions.

For this dataset, the standard deviation is approximately *1.54*, suggesting that most ratings are reasonably close to the average.

5. Histogram

A histogram displays the frequency distribution of customer ratings.

It helps identify:

* Most common ratings
* Spread of responses
* Shape of the distribution
* Possible skewness
* Unusual observations (outliers)


### Preferred summary statistics and visualizations

I would porsonally prefer to use **Histogram** since the survey score are discrete and hence histogram displays the exact distribution of responses across each rating bin. It highlights that responses peak around 7, 8, and 9, showing a strong right-leaning concentration (positive satisfaction).

```python
'''
Part B: Python code to create a histogram using Matplotlib for the survey data:

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
'''

import numpy as np
import matplotlib.pyplot as plt

# Survey scores
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

# Calculate statistics
mean = np.mean(survey_scores)
median = np.median(survey_scores)
std_dev = np.std(survey_scores)
mode = max(set(survey_scores), key=survey_scores.count)

print("Mean:", round(mean,2))
print("Median:", median)
print("Mode:", mode)
print("Standard Deviation:", round(std_dev,2))

# Plot Histogram
plt.figure(figsize=(8,5))
plt.hist(survey_scores,
         bins=7,
         edgecolor='black')

plt.title("Histogram of Customer Satisfaction Scores")
plt.xlabel("Survey Score")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```
