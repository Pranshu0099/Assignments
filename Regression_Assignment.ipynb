**1. What is Simple Linear Regression?**

**Answer:** Simple Linear Regression is a statistical technique used to model the relationship between one independent variable (X) and one dependent variable (Y).

It assumes that the relationship can be represented by a straight line.


> Y = mX + c

Here, **Y** is the predicted dependent variable,
      **X** is the independent variable,
      **m** is the slope/coefficient,
  and **c** is the intercept.
  
For example, salary can be predicted from years of experience.

---

**2. What are the key assumptions of Simple Linear Regression?**

**Answer:** The major assumptions Include:

1. **Linearity:** The relationship between X and Y should be approximately linear.
2. **Independence:** Observations should be independent.
3. **Homoscedasticity:** The variance of errors should remain approximately constant.
4. **Normality of residuals:** Residuals should approximately follow a normal distribution.
5. **No influential outliers:** Extreme observations should not strongly distort the model.
___

**3. What does the coefficient m represent in the equation Y = mX + c?**

**Answer:** The coefficient m represents the slope of the regression line. It indicates how much the predicted value of Y changes when X increases by one unit.
> m = Change in Y / Change in X

For example, if Y = 5000X + 20000,

then an increase of one year in X increases predicted Y by 5,000.

---

**4. What does the intercept c represent in the equation Y = mX + c?**

**Answer:** The intercept c represents the predicted value of Y when X is equal to zero.

>Y = 5000X + 20000  →  when X = 0, Y = 20,000

Therefore, 20,000 is the intercept.

> However, the intercept may not always have a meaningful real world interpretation in X =0 is outside the practicaL range of the data.

**6. What is the purpose of the least squares method in Simple Linear Regression?**

 **Answer:** The least squares method finds the best-fitting regression line by minimizing the sum of squared differences between actual and predicted values.

 > Minimize  Σ(Yi − i)²

 The difference between an actual and predicted value is called a residual or error. Squaring the errors prevents positive and negative errors from cancelling each other.

 ---

**7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?**

**Answer:** R² measures the proportion of variation in the dependent variable that is explained by the independent variable.

>0 ≤ R² ≤ 1

For example, R² = 0.80 means that 80% of the variation in Y is explained by X, while the remaining 20% is due to other factors and random variation.

---

**8. What is Multiple Linear Regression?**

**Answer:** Multiple Linear Regression extends Simple Linear Regression by using two or more independent variables to predict a dependent variable.

> Y = β0 + β1X1 + β2X2 + ... + βnXn + ε

For example, house price can be predicted using area, number of bedrooms, location and age of the house.

---

**9. What is the main difference between Simple and Multiple Linear Regression?**

 **Answer:**Simple Linear Regression uses one independent variable, while Multiple Linear Regression uses two or more independent variables.

 > Simple: Y = mX + c

 > Multiple: Y = β0 + β1X1 + ... + βnXn


 Simple regression is easier to interpret.
 Multiple regression can model the combined effects of several predictors.

---

**10. What are the key assumptions of Multiple Linear Regression?**
**Answer:**The main assumptions are:

(1) linearity,

(2) independence of observations,

(3) homoscedasticity,

(4) approximately normal residuals,

(5) no severe multicollinearity among predictors, and

(6) no highly influential outliers.


---
**11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?**


**Answer:** Heteroscedasticity occurs when the variance of regression errors is not constant across different levels of the predicted variable.

**Example:** residuals are tightly spread for low predicted values but widely spread for high predicted values. It can make standard errors, confidence intervals and hypothesis tests unreliable.

The coefficient estimates may still be unbiased under suitable conditions, but their estimated uncertainty can be incorrect.

---

**12. How can you improve a Multiple Linear Regression model with high multicollinearity?**

**Answer:** Multicollinearity occurs when independent variables are highly correlated. It can be handled by removing one of the correlated variables, combining variables where appropriate, using Principal Component Analysis (PCA), or applying Ridge or Lasso Regression. A common diagnostic is Variance Inflation Factor (VIF). Collecting more relevant data can also help in some situations.

**13. What are some common techniques for transforming categorical variables for use in regression models?**

**Answer:** Categorical variables can be converted into numerical form using techniques such as One-Hot Encoding, Dummy Encoding, and Ordinal Encoding.

> Example: Low = 1, Medium = 2, High = 3 (ordinal encoding)

For nominal categories such as cities, one-hot/dummy encoding is usually appropriate. Ordinal encoding is suitable when categories have a meaningful order.

**14. What is the role of interaction terms in Multiple Linear Regression?** **Answer:** Interaction terms allow a model to represent situations where the effect of one predictor depends on the value of another predictor.

> Y = β0 + β1X1 + β2X2 + β3(X1X2) + ε

Here, X1X2 is the interaction term.

For example, the effect of advertising expenditure on sales may depend on the advertising channel.

**15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?**

**Answer:** In Simple Linear Regression, the intercept is the predicted value of Y when the single predictor is zero.

In Multiple Linear Regression, it is the predicted value of Y when all independent variables are zero. Y = 10 + 2X1 + 5X2 Here, the intercept 10 is the predicted Y when X1 = 0 and X2 = 0.

Its practical meaning depends on whether zero is realistic for all predictors.

---

**16. What is the significance of the slope in regression analysis, and how does it affect predictions?**

**Answer:** The slope represents the expected change in the dependent variable for a one-unit increase in an independent variable, holding other variables constant in Multiple Linear Regression.

> Y = 100 + 20X

A slope of 20 means that a one-unit increase in X is associated with a 20-unit increase in predicted Y.

A positive slope indicates a positive relationship and a negative slope indicates a negative relationship.

**17. How does the intercept in a regression model provide context for the relationship between variables?**

**Answer:** The intercept provides a baseline predicted value of the dependent variable when all predictors are zero.

Salary = 25,000 + 5,000(Experience)
Here, 25,000 is the predicted salary when experience is zero. The intercept is meaningful only when the zero point is relevant to the problem.

**18. What are the limitations of using R² as a sole measure of model performance?**

**Answer:** R² alone is not sufficient because it does not directly measure prediction error, can increase when unnecessary predictors are added, does not verify model assumptions, does not prove causation, and may not indicate how well the model performs on unseen data.

*Useful additional measures:* Adjusted R², RMSE, MAE, cross-validation and residual analysis.

Therefore, model performance should be evaluated using several complementary measures.

---

**19. How would you interpret a large standard error for a regression coefficient?**

**Answer:** A large standard error means that the estimated coefficient has high uncertainty and is less precisely estimated.

Example: β = 10, SE = 8

Possible causes include small sample size, high variability, multicollinearity and a weak relationship between predictor and response. A large standard error can also produce a larger p-value.

---

**20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?**

**Answer:** Plot residuals against fitted or predicted values. A funnel or cone-shaped pattern, where residual spread increases or decreases with fitted values, is a common sign of heteroscedasticity.

> Residuals vs. fitted values → look for changing spread.

It matters because standard errors, confidence intervals and statistical tests can become unreliable.

Possible solutions include transformations, weighted least squares and heteroscedasticity-robust standard errors.

---

**21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?**

**Answer:** It may indicate that some predictors added to the model contribute little useful information.

R² cannot decrease when predictors are added, whereas Adjusted R² penalizes unnecessary predictors.

> High R² + relatively low Adjusted R² → possible unnecessary predictors/overfitting.

This suggests checking the predictors and comparing validation performance before keeping the more complex model.

**22. Why is it important to scale variables in Multiple Linear Regression?**

**Answer:** Scaling puts numerical variables on comparable scales. It is especially important for regularization methods such as Ridge and Lasso because these methods penalize coefficient sizes.

> Standardization: z = (x − µ) / σ

> Min-Max: x' = (x − xmin) / (xmax − xmin)

For ordinary least squares regression, scaling is generally not required for obtaining fitted predictions, although it can improve numerical stability and make coefficient magnitudes easier to compare.

**23. What is polynomial regression?**

**Answer:** Polynomial Regression models a non-linear relationship by including powers of an independent variable.

> Y = β0 + β1X + β2X² + ε

It can model curved relationships rather than only straight-line relationships.

---

**24. How does polynomial regression differ from linear regression?**
**Answer:** Linear Regression models a straight-line relationship, while Polynomial Regression includes higher powers of the predictor and can therefore represent curved relationships.

> Linear: Y = β0 + β1X

> Polynomial: Y = β0 + β1X + β2X² + β3X³ + ...

Polynomial regression is still linear in its coefficients, even though the relationship between X and Y is nonlinear.

---

**25. When is polynomial regression used?**

**Answer:** Polynomial Regression is used when the relationship between the predictor and target is curved rather than approximately linear. Examples: temperature vs. electricity consumption; speed vs. fuel efficiency; experience vs. productivity. It is useful when a straight line cannot adequately represent the observed pattern.

---

**26. What is the general equation for polynomial regression?**
**Answer:** The general equation is:Y = β0 + β1X + β2X² + β3X³ + ... + βnX + ε β0 is the intercept, β1 through βn are coefficients, n is the polynomial degree, and ε is the error term.

---

**27. Can polynomial regression be applied to multiple variables?**

**Answer:** Yes. Polynomial regression can be extended to multiple independent variables by including squared terms, higher powers and, where appropriate, interaction terms.

> Y = β0 + β1X1 + β2X2 + β3X1² + β4X2² + β5X1X2 + ε

However, adding many terms increases complexity and the risk of overfitting and multicollinearity.

---

**28. What are the limitations of polynomial regression?**

**Answer:** Major limitations include

1. overfitting at high degrees,
2. unrealistic extrapolation outside the training range,
3. increased model complexity,
4. multicollinearity among polynomial terms,
5. sensitivity to outliers, and
6. difficulty in selecting the appropriate degree.

---

**29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?**

**Answer:** Useful methods include R², Adjusted R², RMSE, MAE, cross-validation, residual analysis, and information criteria such as AIC or BIC.

Best practice: evaluate performance on validation/test data. The goal is to choose a degree that gives good predictive performance without unnecessary complexity.

---

**30. Why is visualization important in polynomial regression?**

**Answer:** Visualization helps determine whether the relationship is actually nonlinear and whether the chosen polynomial degree is appropriate.

A scatter plot with the fitted polynomial curve can reveal curvature, outliers, underfitting, overfitting and residual patterns.

For example, a degree-2 curve may fit the data well while a degree-10 curve may make unnecessary fluctuations and overfit the data.

**31. How is polynomial regression implemented in Python?**

**Answer:** Polynomial Regression can be implemented using Scikit-learn's PolynomialFeatures and LinearRegression classes.

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)
y_pred = model.predict(X_poly)

PolynomialFeatures(degree=2) creates features such as 1, X and X². The degree can be changed to 3, 4, etc., depending on the required model.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# 1. Define input data (X must be 2D, y can be 1D)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2.5, 4.8, 9.1, 16.2, 25.1])

# 2. Transform features to include x^2 terms
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# 3. Fit linear regression on transformed features
model = LinearRegression()
model.fit(X_poly, y)

# 4. Predict target values
y_pred = model.predict(X_poly)

print("Predictions:", np.round(y_pred, 2))
```
