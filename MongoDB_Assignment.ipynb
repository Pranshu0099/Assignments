# Theoretical Questions

**1. What are the key differences between SQL and NoSQL databases?**

**Answer:**

>Data Model: SQL databases are relational and store structured data in fixed tables with rows and columns. NoSQL databases are non-relational and store unstructured or semi-structured data as documents (JSON/BSON), key-value pairs, wide-column stores, or graphs.

>Schema: SQL requires a predefined, rigid schema. NoSQL uses a dynamic, schema-less (or flexible schema) approach.

>Scaling: SQL typically scales vertically (adding more CPU, RAM, or SSDs to a single server). NoSQL is designed to scale horizontally (distributing data across multiple commodity machines/shards).

>Language & Transactions: SQL uses standard SQL for querying and prioritizes strong ACID guarantees. NoSQL systems use proprietary query APIs (e.g., MQL for MongoDB) and often prioritize high availability and partition tolerance.

**2. What makes MongoDB a good choice for modern applications?**

**Answer:**

> Flexible Document Model: Data is stored in BSON (binary JSON), allowing nested fields and arrays that match modern programming objects naturally without complex Object-Relational Mapping (ORM).

> High Scalability: Built-in horizontal scaling through sharding enables handling massive datasets and traffic spikes with minimal configuration changes.

> High Availability: Automated failover and data redundancy via replica sets keep applications running with minimal downtime.

> Rich Querying & Indexing: Offers extensive indexing, full-text search, geospatial queries, and powerful data aggregation pipelines.

**3. Explain the concept of collections in MongoDB.**

**Answer:** A collection in MongoDB is a grouping of MongoDB documents (analogous to a table in a relational database).

Collections exist within a database and do not enforce a fixed schema by default. This means documents inside the same collection can contain different fields or data types unless explicit validation rules are configured.

**4. How does MongoDB ensure high availability using replication?**

**Answer:**

MongoDB uses Replica Sets, which are clusters of interconnected MongoDB instances (nodes) holding identical copies of the dataset.

If the primary node crashes or becomes unreachable, the remaining secondary nodes hold an automatic election to promote a secondary node as the new primary, preventing service interruptions.

**6. What is the role of indexes in MongoDB, and how do they improve performance?**

**Answer:**

Role: Indexes are specialized data structures (typically B-Trees) that store a small portion of the collection's data in an easily traversable form.

Performance Improvement: Without an index, MongoDB must execute a collection scan (COLLSCAN), inspecting every document to fulfill a query. With an index (IXSCAN), the engine jumps directly to matching records, dramatically reducing disk I/O, memory usage, and execution time.

**7. Describe the stages of the MongoDB aggregation pipeline.**

**ANswer:** An aggregation pipeline processes documents sequentially through multiple transformative stages, where the output of one stage serves as the input to the next:

$match: Filters documents to pass only those matching specific criteria (like a WHERE clause).

$project: Reshapes documents by adding, removing, or renaming fields.

$group: Groups documents by a specified identifier and calculates aggregate metrics (sums, averages, counts).

$sort: Reorders the documents based on specified fields.

$limit / $skip: Restricts the number of output documents or skips a specific count for pagination.

$unwind: Deconstructs an array field from the input documents to output a document for each element.

---

**8. What is sharding in MongoDB? How does it differ from replication?**

**Answer:**

Sharding is the process of partitioning large datasets across multiple servers (shards) to distribute read/write load and increase storage capacity (horizontal scaling).

**Difference from Replication:**

Replication copies the same data across multiple machines to provide redundancy, failover, and high availability.

Sharding splits different portions of data across multiple machines to handle large storage volumes and high throughput.

**10. What are the ACID properties in the context of MongoDB transactions?**

**Answer:** MongoDB provides full multi-document ACID support:

Atomicity: All operations within a transaction succeed together or are rolled back completely ("all or nothing").

Consistency: Data remains valid according to database rules, constraints, and schemas before and after the transaction.

Isolation: Uncommitted changes made inside a transaction remain invisible to other concurrent operations until committed.

Durability: Once committed, all data changes survive unexpected server crashes or network failures (written to write-ahead journals).

**11. What is the purpose of MongoDB’s explain() function?**

**Answer:** explain() is a query profiling tool that returns detailed execution plans and diagnostic statistics for query operations.

It helps developers analyze whether a query used an index (IXSCAN) or scanned the entire database (COLLSCAN), identify performance bottlenecks, and monitor execution times and scanned document counts.

---

**12. How does MongoDB handle schema validation?**

**Answer:** While MongoDB is schema-flexible by default, it supports schema validation using JSON Schema rules applied at the collection level via validator commands.

You can enforce required fields, strict data types (e.g., string, int), allowed value ranges, and regex patterns. MongoDB rejects or warns against insert/update operations that violate these rules based on your validation action settings (error or warn).

---

**13. What is the difference between a primary and a secondary node in a replica set?**

**Answer:** Primary Node: The single master node that receives and processes all write operations and records modifications in its operation log (oplog).

Secondary Node: Replicates the primary's oplog and applies operations asynchronously to maintain an identical copy of the dataset. Secondaries can serve read-only queries when configured and participate in elections to become the new primary if the primary fails.

---

**14. What security mechanisms does MongoDB provide for data protection?**

**Answer:** Authentication: Verifies user identity via SCRAM, x.509 certificates, LDAP, or Kerberos.

Authorization: Employs Role-Based Access Control (RBAC) to restrict user permissions to specific databases or actions.

Encryption: Protects data in transit using TLS/SSL and data at rest using storage encryption algorithms (AES-256) and Queryable Encryption / Client-Side Field Level Encryption (CSFLE).

Auditing & Network Controls: Provides detailed audit logs, IP whitelisting, and VPC peering to control and inspect network traffic.

**15. Explain the concept of embedded documents and when they should be used.**

**Answer:** Concept: Embedding refers to nesting related data structures (sub-documents or arrays of documents) inside a single parent document rather than referencing separate collections.

When to use:

For "contains" or 1-to-1 and 1-to-few relationships (e.g., a user document containing two postal addresses).

When related data is frequently read and updated together, minimizing the need for multi-collection queries.

**16. What is the purpose of MongoDB’s $lookup stage in aggregation?**

**Answer:** The $lookup stage performs a left outer join to combine documents from an external collection into the current aggregation pipeline.

It searches another collection for documents matching a specified key and appends the matched documents as an array inside the output document.

**18. What are the advantages of using MongoDB for horizontal scaling?**

**19. How do MongoDB transactions differ from SQL transactions?**

**Answer:** Usage Scope: SQL databases rely heavily on multi-table transactions for standard operations due to normalized relational schemas. MongoDB's document model handles most updates atomically within a single document without requiring explicit transactions.

Multi-Document Support: SQL natively supports distributed multi-table transactions. MongoDB supports both single-document atomicity (default, zero overhead) and distributed multi-document ACID transactions across replica sets and sharded clusters when complex coordinated writes are necessary.

***

**20. What are the main differences between capped collections and regular collections?**

**Answer:** Capped Collections: Fixed-size, high-throughput collections that work like circular buffers. Once the allocated storage limit is reached, MongoDB automatically overwrites the oldest documents in insertion order. They do not allow document deletion or size-increasing updates.

Regular Collections: Dynamic collections that grow without fixed size constraints, supporting standard CRUD operations (including arbitrary document deletion and flexible resizing).

---

**21. What is the purpose of the $match stage in MongoDB’s aggregation pipeline?**

**Answer:**

The $match stage acts as a filter to pass only documents matching specified query criteria to the subsequent stage of the pipeline.

Best Practice: Placing $match as early as possible in the aggregation pipeline utilizes indexes effectively and minimizes the volume of data passed to subsequent stages.

---

**22. How can you secure access to a MongoDB database?**

**Answer:** Enable Authentication: Require credentials for all database connections (e.g., using auth = true).

Enforce Role-Based Access Control (RBAC): Grant users only the minimum permissions necessary for their tasks (Principle of Least Privilege).

Configure Network Firewalls: Restrict database port access (default: 27017) using IP whitelisting, firewalls, and VPC peering.

Enable Encryption: Enforce TLS/SSL for all network connections and enable encryption at rest for physical storage files.

---

**23. What is MongoDB’s WiredTiger storage engine, and why is it important?**

**Answer:** What it is: WiredTiger is the default, high-performance storage engine in MongoDB responsible for managing how data and indexes are written to and read from disk.

Why it is important:

Document-Level Concurrency: Provides document-level locking, enabling multiple clients to write to different documents in the same collection simultaneously.

Data & Index Compression: Uses compression algorithms (such as Snappy and zlib) to significantly reduce disk storage requirements and reduce I/O bottlenecks.

Checkpointing & Cache Management: Automatically manages memory caches and creates write-ahead journals/checkpoints to ensure high throughput and crash recovery.

```python
import pandas as pd
import pymongo
from pymongo import MongoClient


# Database Connection Setup

client = MongoClient("mongodb://localhost:27017/")
db = client["SuperstoreDB"]
orders = db["Orders"]

# Clear any previous run data
orders.drop()


# Q1: Load the Superstore dataset from CSV into MongoDB

df = pd.read_csv("Sample - Superstore.csv", encoding="windows-1252")
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
df["Profit"] = pd.to_numeric(df["Profit"], errors="coerce")

insert_res = orders.insert_many(df.to_dict(orient="records"))
print(f"--- Q1: Ingestion ---")
print(f"Successfully loaded {len(insert_res.inserted_ids)} records into MongoDB.\n")


# Q2: Retrieve and print all documents from Orders collection

print(f"--- Q2: Retrieve Documents (Displaying first 2) ---")
for doc in orders.find().limit(2):
    print(doc)
print()


# Q3: Count and display total number of documents

total_count = orders.count_documents({})
print(f"--- Q3: Total Document Count ---")
print(f"Total Orders in Collection: {total_count}\n")


# Q4: Fetch all orders from the 'West' region

west_query = {"Region": "West"}
west_count = orders.count_documents(west_query)
print(f"--- Q4: Orders from 'West' Region (Found: {west_count}) ---")
for doc in orders.find(west_query).limit(3):
    print(f"Order ID: {doc.get('Order ID')} | Customer: {doc.get('Customer Name')} | State: {doc.get('State')}")
print()


# Q5: Find orders where Sales > 500

sales_query = {"Sales": {"$gt": 500}}
sales_count = orders.count_documents(sales_query)
print(f"--- Q5: Orders with Sales > $500 (Found: {sales_count}) ---")
for doc in orders.find(sales_query).limit(3):
    print(f"Order ID: {doc.get('Order ID')} | Product: {doc.get('Product Name')[:30]} | Sales: ${doc.get('Sales'):.2f}")
print()


# Q6: Fetch top 3 orders with highest Profit

print(f"--- Q6: Top 3 Most Profitable Orders ---")
top_3_profit = orders.find().sort("Profit", pymongo.DESCENDING).limit(3)
for rank, doc in enumerate(top_3_profit, 1):
    print(f"{rank}. Order ID: {doc.get('Order ID')} | Product: {doc.get('Product Name')} | Profit: ${doc.get('Profit'):.2f}")
print()


# Q7: Update Ship Mode: 'First Class' -> 'Premium Class'

update_res = orders.update_many(
    {"Ship Mode": "First Class"},
    {"$set": {"Ship Mode": "Premium Class"}}
)
print(f"--- Q7: Update Ship Mode ---")
print(f"Matched: {update_res.matched_count} | Modified: {update_res.modified_count} orders updated to 'Premium Class'.\n")


# Q8: Delete all orders where Sales < 50

delete_res = orders.delete_many({"Sales": {"$lt": 50}})
print(f"--- Q8: Delete Low Sales Orders ---")
print(f"Deleted {delete_res.deleted_count} orders where Sales < $50.\n")


# Q9: Aggregate: Group by Region and calculate total sales

print(f"--- Q9: Total Sales per Region ---")
region_pipeline = [
    {
        "$group": {
            "_id": "$Region",
            "total_sales": {"$sum": "$Sales"},
            "order_count": {"$sum": 1}
        }
    },
    {"$sort": {"total_sales": -1}}
]
for row in orders.aggregate(region_pipeline):
    print(f"Region: {row['_id']:<10} | Total Sales: ${row['total_sales']:,.2f} | Remaining Orders: {row['order_count']}")
print()


# Q10: Fetch all distinct values for Ship Mode

distinct_modes = orders.distinct("Ship Mode")
print(f"--- Q10: Distinct Ship Modes ---")
print(f"Ship Modes: {distinct_modes}\n")


# Q11: Count number of orders for each Category

print(f"--- Q11: Order Count by Category ---")
category_pipeline = [
    {
        "$group": {
            "_id": "$Category",
            "total_orders": {"$sum": 1}
        }
    },
    {"$sort": {"total_orders": -1}}
]
for row in orders.aggregate(category_pipeline):
    print(f"Category: {row['_id']:<15} | Orders: {row['total_orders']}")
```
